```python
#Problem 1
'''Generate random samples of size (>=500) from each of the following distributions
Normal
Exponential
Uniform
Poisson'''
import numpy as np 
import pandas as P
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

#Set random seed for for reproducibility 
np.random.seed(51)
size=500

#Normal,Exponential,Uniform,Poisson 
data={
      "Normal":
      np.random.normal(loc=0,scale=1,size=size),
      "Exponential":
      np.random.exponential(scale=1,size=size),
      "Uniform":
      np.random.uniform(low=0,high=1,size=size),
      "Poisson":
      np.random.poisson(lam=5,size=size)
     }
print(data)
```
