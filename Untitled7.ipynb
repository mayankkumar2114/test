```python
import numpy as np
import pandas as pd 
import scipy.stats as stats
import seaborn as sns
np.random.seed(51)
normal_data=np.random.normal(loc=0,scale=1,size=500)
exponential_data=np.random.exponential(scale=1,size=500)
uniform_data=np.random.uniform(low=0,high=1,size=500)
poisson_data=np.random.poisson(lam=5,size=500)
data={'normal':normal_data,
      'exponential':exponential_data,
      'uniform':uniform_data,
      'poisson':poisson_data}

for name, sample in [('Normal', normal_data), 
                   ('Exponential', exponential_data),
                   ('Uniform', uniform_data), 
                   ('Poisson', poisson_data)]:
    print(f"\n--- {name} Distribution ---")
    print(f"Mean     : {np.mean(sample):.4f}")
    print(f"Median   : {np.median(sample):.4f}")
    print(f"Std Dev  : {np.std(sample):.4f}")
    print(f"Min      : {np.min(sample):.4f}")
    print(f"Max      : {np.max(sample):.4f}")
```

```python
#Using pandas 

df=pd.DataFrame(data)
p_summary=df.describe()
print(p_summary)

#Using numpy 
numpy_summary = pd.DataFrame({
    "Mean": np.mean(df, axis=0),
    "Median": np.median(df, axis=0),
    "Variance": np.var(df, axis=0, ddof=1),
    "Std_dev": np.std(df, axis=0, ddof=1),
    "Min": np.min(df, axis=0),
    "Max": np.max(df, axis=0),
    "Skewness": df.skew(),
    "Kurtosis": df.kurt()
})
print(numpy_summary)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
plt.hist(normal_data,bins=20,density=True,
         color='steelblue',edgecolor='black',alpha=0.7)
x=np.linspace(-4,4,100) 
plt.plot(x,stats.norm.pdf(x,0,1),'r-',linewidth=2)
plt.title('for normal distribution')
plt.xlabel('density')

plt.subplot(2,2,2)
plt.hist(exponential_data,bins=20,density=True,color='steelblue',edgecolor='black',alpha=0.7)
```

```python
for name, value in data.items():
    plt.figure() 

    # Using Numpy
    plt.hist(value,bins=30,density=True,alpha=0.3,label="Numpy")

    #Using Seaborn
    sns.histplot(value,bins=25,stat="density",label="Seaborn")
    #Create X values for smooth curve
    x=np.linspace(min(value),max(value),500) 
    #Plot theoretical PDF/PMF
    if name=="Normal":
        plt.plot(x,stats.norm.pdf(x,0,1),label="PDF")
    elif name=="Exponential": 
        plt.plot(x,stats.expon.pdf(x),label="PDF")
    elif name=="Uniform": 
        plt.plot(x,stats.uniform.pdf(x,0,1),label="PDF")
    elif name=="Poisson": 
        plt.plot(x,stats.poisson.pmf(np.round(x),4),label="PMF")
    plt.legend()
    plt.title(name+" Distribution")
    plt.savefig("histda.jpg")
    plt.show()
```

```python
plt.figure(figsize=(12,10))

# Plot 1 - Normal
plt.subplot(2,2,1)
plt.hist(normal_data, bins=30, density=True, color='steelblue', edgecolor='black', alpha=0.7,label='normal')
x = np.linspace(-4, 4, 100)
plt.plot(x, stats.norm.pdf(x, 0, 1), 'r-', linewidth=2)
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
# Plot 2 - Exponential
plt.subplot(2,2,2)
plt.hist(exponential_data, bins=30, density=True, color='orange', edgecolor='black', alpha=0.7,label='exponential')
x2 = np.linspace(0, 8, 100)
plt.plot(x2, stats.expon.pdf(x2, 0, 1), 'r-', linewidth=2)
plt.title('Exponential Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
# Plot 3 - Uniform
plt.subplot(2,2,3)
plt.hist(uniform_data, bins=30, density=True, color='green', edgecolor='black', alpha=0.7,label='uniform')
x3 = np.linspace(0, 1, 100)
plt.plot(x3, stats.uniform.pdf(x3, 0, 1), 'r-', linewidth=2)
plt.title('Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
# Plot 4 - Poisson
plt.subplot(2,2,4)
plt.hist(poisson_data, bins=30, density=True, color='purple', edgecolor='black', alpha=0.7,label='poisson')
x4 = np.arange(0, 15)
plt.plot(x4, stats.poisson.pmf(x4, 3), 'r-o', linewidth=2)
plt.title('Poisson Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.tight_layout()
plt.show()
```
