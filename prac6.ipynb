```python
'''Problem:
Part A: Law of Large Numbers (LLN)
Generate random samples from a Uniform (0,1) distribution.
Consider sample sizes
( n = 10, 50, 100, 500, 1000).
For each sample size, compute the sample mean.
Plot the cumulative mean against sample size.
Part B:  Central Limit Theorem (CLT)
Simulate 1,000 independent samples, each of size (n = 30), from a Gamma distribution with shape = 2 and scale = 3.
Compute the sample mean for each sample.
Plot the histogram of the sample means and comment on its shape.
'''

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(51)

n = [10, 50, 100, 500, 1000]

# Uniform Distribution
print("Sample Means for Uniform(0,1):")
plt.figure(figsize=(10, 6))

for i in n:
    sample = np.random.uniform(0, 1, i)
    mean_i = np.mean(sample)
    print("For n=", i, ": Mean=", mean_i)
    cum_mean = np.cumsum(sample) / np.arange(1, i + 1)
    plt.plot(cum_mean, label=f"n={i}")

plt.axhline(y=0.5, linestyle="--", label="Theoretical Mean=0.5")
plt.xlabel("Sample Size")
plt.ylabel("Cumulative Mean")
plt.title("Cumulative Means for Different Sample Sizes Of Uniform[0,1]")
plt.legend()
plt.savefig("Uniform.jpg")
plt.show()

# Exponential Distribution
print("Sample Means for Exponential(mean=2):")
plt.figure(figsize=(10, 6))

for i in n:
    sample = np.random.exponential(scale=2, size=i)
    mean_i = np.mean(sample)
    print("For n=", i, ": Mean=", mean_i)
    cum_mean = np.cumsum(sample) / np.arange(1, i + 1)
    plt.plot(cum_mean, label=f"n={i}")

plt.axhline(y=2, linestyle="--", label="Theoretical Mean=2")
plt.xlabel("Sample Size")
plt.ylabel("Cumulative Mean")
plt.title("Cumulative Means for Different Sample Sizes Of Exponential(0.5)")
plt.legend()
plt.savefig("Exponential.jpg")
plt.show()

# Central Limit Theorem
shape = 2
scale = 3
n = 30
num_samples = 1000
sample_means = []

for i in range(num_samples):
    sample = np.random.gamma(shape, scale, n)
    sample_means.append(np.mean(sample))

plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=30, edgecolor="black")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("CLT: Sampling Distribution of the Mean (Gamma Distribution)")
plt.savefig("clt.jpg")
plt.show()


'''Interpretation: The simulation verified the Law of Large Numbers and the Central
Limit Theorem. As sample size increased, the sample mean of Uniform (0,1) stabilized
near 0.5, illustrating LLN, while the distribution of means from Gamma samples
appeared approximately normal, demonstrating CLT.'''
```
