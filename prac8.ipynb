```python
import scipy.stats as stats
import numpy as np 
#Case 1
marks=[68,72,69,71,70,67,73,74,66,69]
mu=70
alpha=0.05

'''Parametric Test'''
t_stat,p_val,=stats.ttest_1samp(marks,mu)
print("By t-test for 1 Sample:")
print("Test statistic value:",round(t_stat,4))
print("P-value is:",round(p_val,4))

'''Non-Parametric Test'''
diff=np.array(marks)-mu
w_stat,p_val=stats.wilcoxon(diff)
print("By Wilcoxon-sign Run test for 1 Sample:")
print("Test statistic value:",round(w_stat,4))
print("P-value is:",round(p_val,4))
print("\n")
if p_val>alpha:
    print("The Claim of Coaching Institute is True.","\n","i.e, the average score is 70")
if p_val<alpha:
    print("The Claim of Coaching Institute is False.","\n","i.e, the average score is not 70")
print("\n")

#Case 2 
group_a=[75,78,74,77,76,73,79,80] 
group_b=[72,70,68,71,69,73,67,70] 
m_a=np.mean(group_a) 
m_b=np.mean(group_b) 
std_a=np.std(group_a,ddof=1) 
std_b=np.std(group_b,ddof=1)
n_a=len(group_a) 
n_b=len(group_b) 

'''Parametric Test''' 
t_stat,p_val=stats.ttest_ind_from_stats(m_a,std_a,n_a,m_b,std_b,n_b)
print("By t-test for 2 Independent Sample:")
print("Test statistic value:",round(t_stat,4))
print("P-value is:",round(p_val,4))

'''Non-Parametric Test''' 
m_stat,p_val=stats.mannwhitneyu(group_a,group_b)
print("By Man-Whitney U test for 1 Sample:")
print("Test statistic value:",round(m_stat,4))
print("P-value is:",round(p_val,4))
print("\n") 
if p_val>alpha:
    print("The two teaching methods have same effect.","\n","i.e, there is no significant difference in marks by two different methods")
if p_val<alpha:
    print("The two teaching methods doesn't have same effect.","\n","i.e, there is significant difference in marks by two different methods")
print("\n")

#Case 3 
before=[72,75,68,80,77] 
after=[70,70,62,77,75] 

'''Parametric Test''' 
t_stat,p_val=stats.ttest_rel(before,after)
print("By t-test for 2 Non-Independent Samples:")
print("Test statistic value:",round(t_stat,4))
print("P-value is:",round(p_val,4))

'''Non-Parametric Test''' 
diff=np.array(before)-np.array(after)
w_stat,p_val=stats.wilcoxon(diff,alternative="greater")
print("By Wilcoxon-sign Run test for 2 Non-Independent Samples:")
print("Test statistic value:",round(w_stat,4))
print("P-value is:",round(p_val,4))
print("\n") 
if p_val>alpha:
    print("The Diet Program is not effective.","\n","i.e, there is no significant difference in weights before and after Diet")
if p_val<alpha:
    print("The Diet Program is effective.","\n","i.e, there after Diet are less than that of before Diet")
print("\n")

#Case 4
table=[[30,20],[25,35]] 
'''Test for Independency ''' 
chi_val,p_val,dof,exp=stats.chi2_contingency(table) 
print("Chisquare Test for Independency:")
print("Test statistic value:",round(chi_val,4))
print("P-value is:",round(p_val,4))
print("\n")
if p_val>alpha:
    print("The Course Preference is Independent of Gender")
if p_val<alpha:
    print("The Course Preference is Dependent of Gender")
print("\n")

```
