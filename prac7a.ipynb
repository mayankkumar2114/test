```python
'''Lifetime (hours):

12.4,10.8.11.6,13.1,12.9,11.3,12.0,10.9,11.8,12.5
Assume lifetimes follow N (μ, σ2)
(a)	Obtain MLE of μ and σ2
(b)	Construct 95% confidence interval for μ.
(c)	Use Bootstrap resampling (1000 samples) to estimate the standard error.
(d)	Compare bootstrap and theoretical standard error
(e)	Comment on advantages of bootstrapping'''

import numpy as np
import scipy.stats as stats
from scipy.stats import bootstrap

lt = [12.4, 10.8, 11.6, 13.1, 12.9, 11.3, 12, 10.9, 11.8, 12.5]

# MLE of Population mean is sample mean and population variance is sample variance when X~N(mu,sigma)
m_mle = np.mean(lt)
print("MLE of mean is:", m_mle)

std_mle = np.std(lt, ddof=1)
var_mle = std_mle ** 2
print("MLE of Variance is:", round(var_mle, 4))

# Confidence interval for mu at 95%
ln = len(lt)
df = ln - 1
confd = 0.95
t_critic = stats.t.ppf((1 + confd) / 2, df)
ll = (m_mle - t_critic * (std_mle / ln ** 0.5))
ul = (m_mle + t_critic * (std_mle / ln ** 0.5))
print("The 95% Confidence Interval for mu is: (", round(ll, 4), ",", round(ul, 4), ")")

# Bootstrapped Standard error
data = (np.array([12.4, 10.8, 11.6, 13.1, 12.9, 11.3, 12, 10.9, 11.8, 12.5]),)
res = bootstrap(data, np.mean, confidence_level=0.95, n_resamples=1000)
print("Bootstrapping standard error is:", round(res.standard_error, 4))

# Theoretical Standard error
std_err_t = std_mle / (ln ** 0.5)
print("Theoretical standard error is:", round(std_err_t, 4))

'''Interpretation-Bootstrapping reduced standard error compared to Theoretical standard error. This verifies increase in sample size tends to lesser standard error
Advantages of Bootstrapping:
•	 Non-parametric (Assumption-Free): Unlike traditional tests that assume your data follows a normal (bell-shaped) distribution, bootstrapping relies on the actual observed data. This is critical for skewed data or datasets with heavy tails.
•	Versatility: It can estimate sampling error for almost any statistic—even complex ones like the median, correlation coefficients, or ratios—where standard mathematical formulas do not exist or are extremely difficult to calculate.
•	Effective for Small Samples: When a sample is too small to rely on the Central Limit Theorem (which requires larger samples to approximate normality), bootstrapping provides a more reliable way to estimate variability.
•	Practical & Straightforward: It bypasses the need for complex analytical formulas. By using modern computing power to resample data thousands of times, you can "observe" the sampling distribution directly rather than deriving it theoretically.
•	Robustness to Outliers: While outliers still impact results, bootstrapping often provides a clearer visualization of how much these extreme values affect your final estimate compared to traditional point-estimate methods.
'''
```
