```python
import pandas as pd 
import statsmodels.api as sm
import matplotlib.pyplot as plt 
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
method=["Traditional","Traditional","Traditional","Traditional","Traditional","Online","Online","Online","Online","Online"] 
gender=["Male","Male","Female","Female","Female","Male","Male","Female","Female","Female"]
score=[70,72,75,74,76,78,80,82,81,83] 
data=pd.DataFrame(
    {'method':["Traditional","Traditional","Traditional","Traditional","Traditional","Online","Online","Online","Online","Online"],
     'gender':["Male","Male","Female","Female","Female","Male","Male","Female","Female","Female"],
     'score':[70,72,75,74,76,78,80,82,81,83]})
print(data)
model=ols('score~method*gender',data=data).fit() 
anova_table=sm.stats.anova_lm(model,type=2)
print(anova_table)
means=data.groupby(['gender','method'])['score'].mean().unstack() 
print(means)

means.plot(marker='o')
plt.xlabel('Method') 
plt.ylabel('Mean Score') 
plt.title('Interaction Plot: Gender x Method') 
plt.savefig("inter.jpg") 
plt.show()
```
