```python
#Problem 1 
#a. Store the given data in a python variable
study_hours=[2,5,None,3,6,4,5,7,None,3,4,6,5,7,7,3,4,3,4,5,6]
print(study_hours)

#b.To write user defined function clean_data that removes missing values from the dataset.
def clean_data(data):#We are creating a function named clean_data. The word data is the input it will receive.
    cleaned = []#We make an empty list to store good values.
    for item in data:
        if item is not None:#We check — is this item a real number (not missing)?
            cleaned.append(item)# If yes, add it to our clean list.
    return cleaned #Send the result back to whoever called this function.

#cleaned → this is our list (the empty box we created earlier with cleaned = [])
#.append → this is a built-in Python command that means "add something to the end of a list"
#(item) → this is what we want to add — the current number we're looking at

    
#c.To apply the samwe function for study_hours and store the cleaned data in a new variable. 
data1=clean_data(study_hours)
print(data1)
```

```python
#using for loop compute the square of each element in the list
squares_loop=[]
for i in data1:
    squares_loop.append(i**2)


#using list comprehension compute the squares of each element in a single line of code
square_comprehension=[x**2 for x in data1]

print(square_comprehension,squares_loop,sep='\n')


#Why list comprehension is better:

#Shorter code (1 line vs 3 lines)
#Easier to read once you know the pattern
#Slightly faster for Python to run

#table given read that for more details
```

```python
%%writefile -a summary_basic.py
def median(data):
    cleaned=[]
    for i in data:
        if i is not None:
            cleaned.append(i)
    sorted_data=sorted(cleaned)
    n=len(sorted_data)
    middle=n//2
    if n%2==1:
        return sorted_data[middle]
    else:
        return (sorted_data[middle-1]+sorted_data[middle])/2
#this is the function of median

def mode(data):
    cleaned=[]
    for i in data:
        if i is not None:
            cleaned.append(i)
    frequency={}
    for x in cleaned:
        if x in frequency:
            frequency[x]+=1
        else:
            frequency[x]=1
    return max(frequency,key=frequency.get)
    #this is the function of mode

def standarddev(data):
    cleaned=[]
    for i in data:
        if i is not None:
            cleaned.append(i)
    avg=mean(cleaned)
    squared_diffs=[]
    for x in cleaned:
        squared_diffs.append((x-avg)**2)
    variance=sum(squared_diffs)/(len(cleaned)-1)
    return variance**0.5
#this is the standard deviation
```
