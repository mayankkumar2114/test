```python
study_hours=[2,4,5,None,3,6,4,5,7,None,3,4,6,5,7,7,3,4,3,4,5,6]
#problem1
def clean1(data):
    return [item for item in data if item is not None and str(item).strip()!=""]
data1=clean1(study_hours)
print(data1)

def clean_data(data):
    cleaned=[]
    for item in data:
        if item is not None:
            cleaned.append(item)
    return cleaned
data2=clean_data(study_hours)
print(data2)

#problem2
sq=[]
for i in data2:
    sq.append(i**2)
print(sq)
sq1=[i**2 for i in data2]
print(sq1) 
print('part1:',sq,'\npart2:',sq1)

#Why list comprehension is better:

#Shorter code (1 line vs 3 lines)
#Easier to read once you know the pattern
#Slightly faster for Python to run

def mean(data):
    cleaned=[]
    for i in data:
        if i is not None:
            cleaned.append(i)
    if len(cleaned)==0:
        return 0
    return sum(cleaned)/len(cleaned)

print('mean of the cleaned study hours:',mean(study_hours))
```
