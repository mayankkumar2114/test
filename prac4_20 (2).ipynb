```python
import pandas as pd 
import numpy as np 
data={'X':[4,6,8,5,7],
      'Y':[7,9,10,8,11]}
df=pd.DataFrame(data)
df.to_csv('bivariate_data.csv',index=False)
print(df.head())


df=pd.read_csv('bivariate_data.csv')
print('Dataframe\n',df)
array=df.to_numpy()
print('numpy Array:\n',array)

X=array[:,0]
Y=array[:,1]
print(X,Y)


A = array
A_T = A.T
print("\nTranspose of Data Matrix:")
print(A_T)
if A.shape == A_T.shape:
    print("\nA + A^T:")
    print(A + A_T)
else:
    print("\nMatrix addition A + A^T is not defined (different shapes).")
ATA = A_T @ A
print("\nMatrix Multiplication (A^T A):")
print(ATA)
try:
    ATA_inv = np.linalg.inv(ATA)
    print("\nInverse of (A^T A):")
    print(ATA_inv)
except np.linalg.LinAlgError:
    print("\nInverse does not exist.")
print("\nRow sums of A:", A.sum(axis=1))
print("Column sums of A:", A.sum(axis=0))
eigvals, eigvecs = np.linalg.eig(ATA)
print("\nEigenvalues:")
print(eigvals)
print("Eigenvectors:")
print(eigvecs)

mean_X = np.mean(X)
mean_Y = np.mean(Y)
print("\nSample Mean of X:", mean_X)
print("Sample Mean of Y:", mean_Y)
cov_matrix = np.cov(X, Y)
print("\nVariance-Covariance Matrix:")
print(cov_matrix)

print('''(a) Interpret the Covariance value:
The covariance between X and Y will be a positive number. This means when X increases, Y also tends to increase. Both variables move in the same direction.
(b) Relationship between X and Y:
Looking at the data — as X goes from 4→6→8, Y goes from 7→9→10. There is a clear positive relationship — students with higher X values also have higher Y values. They are positively correlated.
(c) Role of Eigenvalues and Eigenvectors in multivariate analysis:
Think of data as a cloud of points in space:

Eigenvalues → tell you how much variation/spread exists in each direction. Larger eigenvalue = more important direction
Eigenvectors → tell you which direction that variation goes
Together they help find the most important patterns in data — this is the foundation of a technique called PCA (Principal Component Analysis) which simplifies complex data while keeping the important information''')
print('''\nFrom the given data, as the values of Xincrease, the corresponding values of Yalso tend to increase. This indicates a positive relationship between Xand Y. In other words, higher values of Xare associated with higher values of Y, suggesting a positive correlation between the two variables. This pattern implies that the variables move in the same direction and are linearly related to some extent.
In multivariate analysis, eigenvalues and eigenvectors describe the main patterns of variation in a dataset. Eigenvalues indicate how much variance is captured in each principal direction, while eigenvectors give the directions (linear combinations of variables) along which this variance is maximized. They form the basis of techniques like Principal Component Analysis (PCA), helping reduce dimensionality, identify important features, and reveal the underlying structure of multivariate data.
''')







#practical5

#Problem 1
'''Generate random samples of size (>=500) from each of the following distributions
Normal
Exponential
Uniform
Poisson'''
import numpy as np 
import pandas as P
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

#Set random seed for for reproducibility 
np.random.seed(51)
size=500

#Normal,Exponential,Uniform,Poisson 
data={
      "Normal":
      np.random.normal(loc=0,scale=1,size=size),
      "Exponential":
      np.random.exponential(scale=1,size=size),
      "Uniform":
      np.random.uniform(low=0,high=1,size=size),
      "Poisson":
      np.random.poisson(lam=5,size=size)
     }
'''To Perform Descriptive analysis of each data'''

#Using pandas 
import pandas as pd
df=pd.DataFrame(data)
p_summary=df.describe()
print(p_summary)

#Using numpy 
numpy_summary = pd.DataFrame({
    "Mean": np.mean(df, axis=0),
    "Median": np.median(df, axis=0),
    "Variance": np.var(df, axis=0, ddof=1),
    "Std_dev": np.std(df, axis=0, ddof=1),
    "Min": np.min(df, axis=0),
    "Max": np.max(df, axis=0),
    "Skewness": df.skew(),
    "Kurtosis": df.kurt()
})
print(numpy_summary)
for name, value in data.items():
    plt.figure() 

    # Using Numpy
    plt.hist(value,bins=30,density=True,alpha=0.3,label="Numpy")

    #Using Seaborn
    sns.histplot(value,bins=25,stat="density",label="Seaborn")
    #Create X values for smooth curve
    x=np.linspace(min(value),max(value),500) 
    #Plot theoretical PDF/PMF
    if name=="Normal":
        plt.plot(x,stats.norm.pdf(x,0,1),label="PDF")
    elif name=="Exponential": 
        plt.plot(x,stats.expon.pdf(x),label="PDF")
    elif name=="Uniform": 
        plt.plot(x,stats.uniform.pdf(x,0,1),label="PDF")
    elif name=="Poisson": 
        plt.plot(x,stats.poisson.pmf(np.round(x),4),label="PMF")
    plt.legend()
    plt.title(name+" Distribution")
    plt.savefig("histda.jpg")
    plt.show()

plt.figure(figsize=(12, 10))
i = 1
for name, data in data.items():
    plt.subplot(2, 2, i)
    stats.probplot(data, dist="norm", plot=plt)
    plt.title(f"Q-Q Plot : {name}")
    i += 1
plt.tight_layout()
plt.show()
print('''Q-Q Plot Interpretation:
1) If points fall close to a straight line => data is approximately Normal.
2) Symmetric distribution => points are balanced around the line.
3) Right-skewed data (Exponential) => points curve away from the line (especially in upper tail).
4) Heavy-tailed data => ends deviate strongly from the line (extreme values present).
5) Light-tailed data => ends bend inward compared to Normal.
6) Poisson/Uniform are not Normal, so Q-Q plot will show systematic deviation from straight line.

''')

'''Problem 6: 
Introduce standardization or transformation (e.g., log or square-root) for a skewed distribution and re-examine normality using a Q–Q plot.'''

'Code:'
df["log_transformed"]=np.log(df["Exponential"])
df["Standardised"]=(df["log_transformed"]-df["log_transformed"].mean())/df["log_transformed"].std() 
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
stats.probplot(df["Exponential"],dist="norm",plot=plt)
plt.title("Q-Q Plot Original Exponential Data")
plt.subplot(1,2,2) 
stats.probplot(df["Standardised"],dist="norm",plot=plt) 
plt.title("Q-Q Plot Log-Transformed & Standardised Exponential Data")
plt.tight_layout()
plt.savefig("Std_sub.jpg")
plt.show()

print('''Interpretation:
	Random samples from Normal, Exponential, Uniform and Poisson distribution were analysed using descriptive statistics, histograms and Q-Q Plots.
    The normal distribution showed symmetry and strong alignment with Q-Q reference line, confirming normality, while Uniform and Poisson distribution
    showed deviations due to non-normal and discrete behaviour. The exponential distribution exhibited strong right skewedness and heavy tails,
    clearly reflected in its Q-Q plot. Overall, symmetric distributions align better with the Q-Q line, whereas skewed distributions show curvature
    and tail deviations. Apply a log transformation to the exponential data reduced skewedness and improved normality, as observed from the Q-Q plot.
    ''')






#practical6
'''Problem:
Part A: Law of Large Numbers (LLN)
Generate random samples from a Uniform (0,1) distribution.
Consider sample sizes
( n = 10, 50, 100, 500, 1000).
For each sample size, compute the sample mean.
Plot the cumulative mean against sample size.
Part B:  Central Limit Theorem (CLT)
Simulate 1,000 independent samples, each of size (n = 30), from a Gamma distribution with shape = 2 and scale = 3.
Compute the sample mean for each sample.
Plot the histogram of the sample means and comment on its shape.
'''

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(51)

n = [10, 50, 100, 500, 1000]

# Uniform Distribution
print("Sample Means for Uniform(0,1):")
plt.figure(figsize=(10, 6))

for i in n:
    sample = np.random.uniform(0, 1, i)
    mean_i = np.mean(sample)
    print("For n=", i, ": Mean=", mean_i)
    cum_mean = np.cumsum(sample) / np.arange(1, i + 1)
    plt.plot(cum_mean, label=f"n={i}")

plt.axhline(y=0.5, linestyle="--", label="Theoretical Mean=0.5")
plt.xlabel("Sample Size")
plt.ylabel("Cumulative Mean")
plt.title("Cumulative Means for Different Sample Sizes Of Uniform[0,1]")
plt.legend()
plt.savefig("Uniform.jpg")
plt.show()

# Exponential Distribution
print("Sample Means for Exponential(mean=2):")
plt.figure(figsize=(10, 6))

for i in n:
    sample = np.random.exponential(scale=2, size=i)
    mean_i = np.mean(sample)
    print("For n=", i, ": Mean=", mean_i)
    cum_mean = np.cumsum(sample) / np.arange(1, i + 1)
    plt.plot(cum_mean, label=f"n={i}")

plt.axhline(y=2, linestyle="--", label="Theoretical Mean=2")
plt.xlabel("Sample Size")
plt.ylabel("Cumulative Mean")
plt.title("Cumulative Means for Different Sample Sizes Of Exponential(0.5)")
plt.legend()
plt.savefig("Exponential.jpg")
plt.show()

# Central Limit Theorem
shape = 2
scale = 3
n = 30
num_samples = 1000
sample_means = []

for i in range(num_samples):
    sample = np.random.gamma(shape, scale, n)
    sample_means.append(np.mean(sample))

plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=30, edgecolor="black")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("CLT: Sampling Distribution of the Mean (Gamma Distribution)")
plt.savefig("clt.jpg")
plt.show()


print('''Interpretation: The simulation verified the Law of Large Numbers and the Central
Limit Theorem. As sample size increased, the sample mean of Uniform (0,1) stabilized
near 0.5, illustrating LLN, while the distribution of means from Gamma samples
appeared approximately normal, demonstrating CLT.''')







#practical7
'''Lifetime (hours):

12.4,10.8.11.6,13.1,12.9,11.3,12.0,10.9,11.8,12.5
Assume lifetimes follow N (μ, σ2)
(a)	Obtain MLE of μ and σ2
(b)	Construct 95% confidence interval for μ.
(c)	Use Bootstrap resampling (1000 samples) to estimate the standard error.
(d)	Compare bootstrap and theoretical standard error
(e)	Comment on advantages of bootstrapping'''

import numpy as np
import scipy.stats as stats
from scipy.stats import bootstrap

lt = [12.4, 10.8, 11.6, 13.1, 12.9, 11.3, 12, 10.9, 11.8, 12.5]

# MLE of Population mean is sample mean and population variance is sample variance when X~N(mu,sigma)
m_mle = np.mean(lt)
print("MLE of mean is:", m_mle)

std_mle = np.std(lt, ddof=1)
var_mle = std_mle ** 2
print("MLE of Variance is:", round(var_mle, 4))

# Confidence interval for mu at 95%
ln = len(lt)
df = ln - 1
confd = 0.95
t_critic = stats.t.ppf((1 + confd) / 2, df)
ll = (m_mle - t_critic * (std_mle / ln ** 0.5))
ul = (m_mle + t_critic * (std_mle / ln ** 0.5))
print("The 95% Confidence Interval for mu is: (", round(ll, 4), ",", round(ul, 4), ")")

# Bootstrapped Standard error
data = (np.array([12.4, 10.8, 11.6, 13.1, 12.9, 11.3, 12, 10.9, 11.8, 12.5]),)
res = bootstrap(data, np.mean, confidence_level=0.95, n_resamples=1000)
print("Bootstrapping standard error is:", round(res.standard_error, 4))

# Theoretical Standard error
std_err_t = std_mle / (ln ** 0.5)
print("Theoretical standard error is:", round(std_err_t, 4))

print('''Interpretation-Bootstrapping reduced standard error compared to Theoretical standard error. This verifies increase in sample size tends to lesser standard error
Advantages of Bootstrapping:
•	 Non-parametric (Assumption-Free): Unlike traditional tests that assume your data follows a normal (bell-shaped) distribution, bootstrapping relies on the actual observed data. This is critical for skewed data or datasets with heavy tails.
•	Versatility: It can estimate sampling error for almost any statistic—even complex ones like the median, correlation coefficients, or ratios—where standard mathematical formulas do not exist or are extremely difficult to calculate.
•	Effective for Small Samples: When a sample is too small to rely on the Central Limit Theorem (which requires larger samples to approximate normality), bootstrapping provides a more reliable way to estimate variability.
•	Practical & Straightforward: It bypasses the need for complex analytical formulas. By using modern computing power to resample data thousands of times, you can "observe" the sampling distribution directly rather than deriving it theoretically.
•	Robustness to Outliers: While outliers still impact results, bootstrapping often provides a clearer visualization of how much these extreme values affect your final estimate compared to traditional point-estimate methods.
''')






#practical8
print('''Methodology:
1. One-Sample t-Test:-
Test Statistic: t= (((x) ̅-μ_0))/(s/√n)
Where,
	x ̅ = sample mean
	s = sample standard deviation
	n = sample size
	μ0=Population Mean

2. Independent Samples t-Test:-
Test Statistic: t= ((x ̅-y ̅))/(√((s_1^2)/n_1 +(s_2^2)/n_2 ))
Where,
	x ̅ = sample mean of x
	y ̅  = sample mean of y
	S1 = sample variance of x
	S2=sample variance of y
	n1=sample size of x
	n2=sample size of y

3. Paired t-Test:-
Test Statistic: t= ((d ̅))/(s_d/√n)
Where,
	d ̅ = mean of differences
	Sd = sample standard deviation
	n = sample size

4. Wilcoxon Signed-Rank Test:- 
     a non-parametric test used to compare two related groups or test whether a single sample median differs from a
     hypothesized value, without assuming normality.
	Null hypothesis (H₀) — the median difference between pairs is zero; no effect exists.
    Alternative hypothesis (H₁) — the median difference is not zero (two-tailed), or is positive/negative (one-tailed).
    Differences — subtract one paired value from the other: d = x₁ − x₂ for each pair.
    Absolute differences — take |d|; the sign is set aside temporarily.
    Zero differences — pairs where d = 0 are discarded from the analysis.
    Ranking — arrange the absolute differences from smallest to largest and assign ranks 1, 2, 3…
    Tied ranks — when two differences are equal, assign the average of the ranks they would have occupied.
    Signed ranks — restore the original sign (+ or −) back to each rank.
    W⁺ — sum of ranks where the difference was positive.
    W⁻ — sum of ranks where the difference was negative.
    Test statistic W — the smaller of W⁺ and W⁻; compared against a critical value from the Wilcoxon table.
    Critical value — a threshold from the Wilcoxon table; if W ≤ critical value, reject H₀.
    p-value — probability of observing a result this extreme if H₀ is true; reject H₀ if p < 0.05.
5. Mann-Whitney U Test:-
	a non-parametric test used to compare two independent groups to see if one tends to have higher values
    than the other, without assuming normality.
    Null hypothesis (H₀) — the two populations have the same distribution; neither group tends to have larger values.
    Alternative hypothesis (H₁) — one population tends to have larger values than the other.
    Ranking — combine both groups into one list and rank all values together from smallest to largest.
    Tied ranks — when two values are equal across groups, assign the average of the ranks they would occupy.
    R₁ — sum of ranks belonging to Group 1.
    R₂ — sum of ranks belonging to Group 2.
    U statistic — measures how often a value from one group exceeds a value from the other group.
    U₁ — U statistic for Group 1 = n₁n₂ + n₁(n₁+1)/2 − R₁
    U₂ — U statistic for Group 2 = n₁n₂ + n₂(n₂+1)/2 − R₂
    Test statistic U — the smaller of U₁ and U₂; compared against a critical value.
    Check — U₁ + U₂ must always equal n₁ × n₂; use this to verify your calculation.
    Critical value — threshold from the Mann-Whitney table; reject H₀ if U ≤ critical value.
    p-value — probability of getting this result by chance if H₀ is true; reject H₀ if p < 0.05.
    Large sample approximation — when n₁ and n₂ are both large (>20), U is converted to a Z-score and compared to the normal distribution.
    6. Chi-Square Test for Independency: -

χ² = Σ [(Oi – Ei) ² / Ei]
where O = observed frequency, E = expected frequency
Expected frequency: E = (Row Total × Column Total) / Grand Total
Degrees of freedom: df = (rows - 1) × (columns - 1) = (2-1) × (2-1) = 1


Rejection Criterion:
1. α approach: -
We Reject Null Hypothesis if Statistic calculated value if |t cal |> t tab.
Where t tab is table value of statistics at α level of significance.
2. P-value approach: - 
	We Reject Null Hypothesis if Statistic calculated value if p-value < α
''')
import scipy.stats as stats
import numpy as np

alpha = 0.05

# Case 1
marks = [68, 72, 69, 71, 70, 67, 73, 74, 66, 69]
mu = 70

print("=" * 50)
print("CASE 1: One Sample Test")
print("=" * 50)

# Parametric Test
t_stat, p_val = stats.ttest_1samp(marks, mu)
print("By t-test for 1 Sample:")
print("Test statistic value:", round(t_stat, 4))
print("P-value is:", round(p_val, 4))

# Non-Parametric Test
diff = np.array(marks) - mu
w_stat, p_val = stats.wilcoxon(diff)
print("By Wilcoxon Signed-Rank test for 1 Sample:")
print("Test statistic value:", round(w_stat, 4))
print("P-value is:", round(p_val, 4))

if p_val > alpha:
    print("Conclusion: The Claim of Coaching Institute is True.")
    print("i.e, the average score is 70\n")
else:
    print("Conclusion: The Claim of Coaching Institute is False.")
    print("i.e, the average score is not 70\n")


# Case 2
print("=" * 50)
print("CASE 2: Two Independent Samples Test")
print("=" * 50)

group_a = [75, 78, 74, 77, 76, 73, 79, 80]
group_b = [72, 70, 68, 71, 69, 73, 67, 70]

m_a = np.mean(group_a)
m_b = np.mean(group_b)
std_a = np.std(group_a, ddof=1)
std_b = np.std(group_b, ddof=1)
n_a = len(group_a)
n_b = len(group_b)

# Parametric Test
t_stat, p_val = stats.ttest_ind_from_stats(m_a, std_a, n_a, m_b, std_b, n_b)
print("By t-test for 2 Independent Samples:")
print("Test statistic value:", round(t_stat, 4))
print("P-value is:", round(p_val, 4))

# Non-Parametric Test
m_stat, p_val = stats.mannwhitneyu(group_a, group_b)
print("By Mann-Whitney U test:")
print("Test statistic value:", round(m_stat, 4))
print("P-value is:", round(p_val, 4))

if p_val > alpha:
    print("Conclusion: The two teaching methods have the same effect.")
    print("i.e, no significant difference in marks\n")
else:
    print("Conclusion: The two teaching methods have different effects.")
    print("i.e, significant difference in marks\n")


# Case 3
print("=" * 50)
print("CASE 3: Paired Samples Test")
print("=" * 50)

before = [72, 75, 68, 80, 77]
after  = [70, 70, 62, 77, 75]

# Parametric Test
t_stat, p_val = stats.ttest_rel(before, after, alternative="greater")
print("By Paired t-test (one-tailed):")
print("Test statistic value:", round(t_stat, 4))
print("P-value is:", round(p_val, 4))

# Non-Parametric Test
diff = np.array(before) - np.array(after)
w_stat, p_val = stats.wilcoxon(diff, alternative="greater")
print("By Wilcoxon Signed-Rank test (Paired):")
print("Test statistic value:", round(w_stat, 4))
print("P-value is:", round(p_val, 4))

if p_val > alpha:
    print("Conclusion: The Diet Program is NOT effective.")
    print("i.e, no significant difference in weights\n")
else:
    print("Conclusion: The Diet Program is effective.")
    print("i.e, weights after diet are significantly less\n")


# Case 4
print("=" * 50)
print("CASE 4: Chi-Square Test of Independence")
print("=" * 50)

table = [[30, 20], [25, 35]]

chi_val, p_val, dof, exp = stats.chi2_contingency(table)
print("Chi-Square Test for Independence:")
print("Test statistic value:", round(chi_val, 4))
print("P-value is:", round(p_val, 4))
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", exp)

if p_val > alpha:
    print("Conclusion: Course Preference is Independent of Gender\n")
else:
    print("Conclusion: Course Preference is Dependent on Gender\n")

print('''Interpretation:
The following Interpretations are drawn after testing with appropriate tests
Case 1: The average score is 70
Case 2: There is significant difference in marks by two different methods
Case 3: There after Diet Weights are less than that of before Diet
Case 4: The Course Preference is Independent of Gender''')






#practical9
print('''Problem: Hypothesis Testing
One – Way ANOVA, F-test & Kruskal -Wallis Test 
A Researcher studies the effect of four treatments on a quantitative response variable.
Use the following data
	G1=42,45,44,43
	G2=48,50,49,47
	G3=46,45,47,44
	G4=52,51,53,50
OR
Generate a reproducible dataset in python for four groups of equal size with approximate means 45, 48, 46 & 52.
	Using the same dataset throughout, perform a one – way ANOVA (F-Test) and interpret the result.
	Check the model assumptions using appropriate diagnostic plots.
	Apply the Kruskal –Wallis test and compare its conclusion with ANOVA.
	Carry out Tukey’s HSD post-hoc test and comment on the statistical and practical significance on the findings.

Methodology:
1. One-Way ANOVA (F-Test)
Objective:
To test whether there is a significant difference among the means of four independent groups.
Hypotheses:
	Null Hypothesis (H₀): μ₁ = μ₂ = μ₃ = μ₄
(All group means are equal)
	Alternative Hypothesis (H₁):
At least one group mean is different.
Test Statistic:
F="Between Group Variance" /"Within Group Variance" 

Where,
	Between-group variance measures variation of group means from overall mean.
	Within-group variance measures variation inside each group.
Decision Rule:
	Reject H₀ if:
	FCal > Ftab
OR
	p-value < α (0.05)

2. Model Assumptions for ANOVA
Before applying ANOVA, the following assumptions must be checked:
(a) Normality
	Tested using Shapiro–Wilk Test
	H₀: Data follows normal distribution
	Reject if p-value < 0.05
(b) Homogeneity of Variances
	Tested using Bartlett’s Test
	H₀: All group variances are equal
	Reject if p-value < 0.05
(c) Independence of Observations
	Verified using:
	Study design
	Residual plot
	Durbin–Watson statistic
If assumptions are satisfied → Proceed with ANOVA.

3. Kruskal–Wallis Test (Non-Parametric Alternative)
Objective: To test whether group medians differ when ANOVA assumptions are violated.
Hypotheses:
	H₀: All group distributions are identical
	H₁: At least one group differs
Test Statistic (H): Based on ranking all observations together.
Decision Rule:
	Reject H₀ if p-value < 0.05

4. Tukey’s HSD Post-Hoc Test
If ANOVA shows significant difference:
	Perform Tukey’s Honest Significant Difference test.
	It identifies which specific pairs of groups differ.
	Controls family-wise error rate.
Decision Rule:
Reject pairwise equality if:
	Adjusted p-value < 0.05
OR
	Confidence interval does not contain zero.

Rejection Criterion:
1. α-Level Approach
Reject H₀ if calculated statistic exceeds critical value.

2. P-Value Approach
Reject H₀ if:
p-value<0.05
''')
import numpy as np
import pandas as pd
import scipy.stats as stats
from statsmodels.stats.stattools import durbin_watson
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import matplotlib.pyplot as plt

G1=np.array([42,45,44,43])
G2=np.array([48,50,49,47])
G3=np.array([46,45,47,44])
G4=np.array([52,51,53,50])

#Shapiro-Wilk test for Normality
groups=[G1,G2,G3,G4]
for i in range(4):
    stat,p=stats.shapiro(groups[i])
    print("Group",i+1,"p-value is:",round(p,4))

#Bartletts test for homogenity of variances
stat,p=stats.bartlett(G1,G2,G3,G4)
print("\n Bartlett's test p-value is:",round(p,4))

#Independency of Observations
residuals=np.concatenate([g-np.mean(g) for g in groups])
dw=durbin_watson(residuals)
print("\n Durbin-Watson test statistic is:",dw)
group_labels=np.concatenate([[i+1]*len(g) for i,g in enumerate(groups)]) 

plt.scatter(group_labels,residuals,color="black")
plt.axhline(0,color="blue",linestyle="--")
plt.xlabel("Group") 
plt.ylabel("Residuals") 
plt.title("Residuals by Group") 
plt.savefig("res1.jpg") 
plt.show() 


#Kruskal-Wallis Test
stat,p=stats.kruskal(G1,G2,G3,G4)
print("\n Kruskal-Wallis test p-value is:",round(p,4))

#One Way ANOVA
f_stat,p=stats.f_oneway(G1,G2,G3,G4)
print("\n One Way ANOVA test p-value is:",round(p,4))


# Tukey HSD Post-hoc Test
values = np.concatenate([G1, G2, G3, G4])
group = (["G1"]*len(G1) + ["G2"]*len(G2) +
          ["G3"]*len(G3) + ["G4"]*len(G4))

df = pd.DataFrame({"value": values, "group": group})
tukey = pairwise_tukeyhsd(endog=df["value"],
                          groups=df["group"],
                          alpha=0.05)
print("\nTukey HSD Post-hoc Results:",tukey)
tukey.plot_simultaneous()
plt.title("Tukey HSD Confidence Intervals")
plt.savefig("hsd1.jpg") 
plt.show()

#For Simulated reproducible random numbers
print("For the Generated Random Numbers:")
np.random.seed(42)
n=100
G1=np.random.normal(45,5,n)
G2=np.random.normal(48,5,n)
G3=np.random.normal(46,5,n)
G4=np.random.normal(52,5,n)

#Kruskal-Wallis Test
stat,p=stats.kruskal(G1,G2,G3,G4)
print("\n Kruskal-Wallis test p-value is:",round(p,4))

#One Way ANOVA
f_stat,p=stats.f_oneway(G1,G2,G3,G4)
print("\n One Way ANOVA test p-value is:",round(p,4))

# Tukey HSD Post-hoc Test
values = np.concatenate([G1, G2, G3, G4])
group = (["G1"]*len(G1) + ["G2"]*len(G2) +
          ["G3"]*len(G3) + ["G4"]*len(G4))

df = pd.DataFrame({"value": values, "group": group})
tukey = pairwise_tukeyhsd(endog=df["value"],
                          groups=df["group"],
                          alpha=0.05)
print("\nTukey HSD Post-hoc Results:",tukey)
tukey.plot_simultaneous()
plt.title("Tukey HSD Confidence Intervals")
plt.savefig("hsd2.jpg")
plt.show() 
print("\n One Way ANOVA test p-value is:",round(p,4))

print('''The following interpretations are drawn after performing the appropriate tests:

•	Assumption Checking:
Shapiro–Wilk test showed p-values greater than 0.05 for all groups, indicating normality.
Bartlett’s test p-value was greater than 0.05, indicating equal variances.
Durbin–Watson statistic was approximately 2, showing independence of observations.
Hence, all ANOVA assumptions are satisfied.
•	One-Way ANOVA:
Since the p-value is less than 0.05, we reject the null hypothesis.
There is a significant difference among the four treatment means.
•	Kruskal–Wallis Test:
The p-value is less than 0.05.
This confirms that at least one group differs significantly.
Thus, both parametric and non-parametric tests give the same conclusion.
•	Tukey’s HSD Test:
Most pairwise comparisons show significant differences.
Group G4 has the highest mean and differs significantly from other groups.
•	Final Conclusion:
There is a statistically significant difference among the four treatments.
Treatment G4 appears to be the most effective among all groups.
''')






#practical10
print('''Problem: Two-Way ANOVA with Interaction Effects
A study examines the effect of Teaching Method and Gender on test scores.
The following data were collected:
1.Create an appropriate Pandas Data Frame for the above data.
2.Perform a two-way ANOVA including interaction effects.
3.Test the significance of: 
(i)	Teaching Method, 
(ii)	Gender, 
(iii)	Method × Gender interaction
4.Draw interaction plots to visualize the effects.
5.Interpret the interaction effect in statistical terms and summarize conclusions.

Methodology:
Two-Way ANOVA: Tests the effect of two categorical factors (Teaching Method and Gender) on a continuous outcome (Score). Partitions total variance into main effects, interaction effect, and residual error.
Main Effects: Teaching Method effect (averaged across genders) and Gender effect (averaged across teaching methods).
Interaction Effect: Tests whether the effect of one factor depends on the level of the other factor. Non-parallel lines in plots indicate significant interaction.
Hypotheses: H₀ (Main effects): Factor has no effect on Score. H₀ (Interaction): No interaction between factors. Rejected if p<0.05.

''')

import pandas as pd 
import statsmodels.api as sm
import matplotlib.pyplot as plt 
from statsmodels.formula.api import ols
method=["Traditional","Traditional","Traditional","Traditional","Traditional","Online","Online","Online","Online","Online"] 
gender=["Male","Male","Female","Female","Female","Male","Male","Female","Female","Female"]
score=[70,72,75,74,76,78,80,82,81,83] 

data=pd.DataFrame(
    {'method':["Traditional","Traditional","Traditional","Traditional","Traditional","Online","Online","Online","Online","Online"],
     'gender':["Male","Male","Female","Female","Female","Male","Male","Female","Female","Female"],
     'score':[70,72,75,74,76,78,80,82,81,83]})
print(data)
model=ols('score~method*gender',data=data).fit() 
anova_table=sm.stats.anova_lm(model,type=2)
print(anova_table)
means=data.groupby(['gender','method'])['score'].mean().unstack() 
print(means)

means.plot(marker='o')
plt.xlabel('Method') 
plt.ylabel('Mean Score') 
plt.title('Interaction Plot: Gender x Method') 
plt.savefig("inter.jpg") 
plt.show()


print('''Interpretation:
•	Main Effects: Teaching Method is the dominant factor (F=102.68, p<0.001). Online teaching significantly outperforms traditional teaching with an average improvement of 7.4 points. Gender also significantly affects scores (F=22.05, p=0.003), with females scoring 3.5 points higher on average.
•	Interaction Effect: The non-significant interaction (F=0.45, p=0.527) indicates that the advantage of Online teaching over Traditional teaching is similar for both males and females. Similarly, the female advantage over males is consistent across both teaching methods. The approximately parallel lines in the interaction plots visually confirm this finding.
•	Practical Significance: Both factors independently improve test scores. For best outcomes, Online teaching should be adopted, and teaching strategies should be gender-neutral as interaction is absent. The main effects are additive - combining online teaching with any gender produces predictable results without unexpected interactions.
•	Recommendations: 
(1) Implement online teaching methods for improved learning outcomes. 
(2) Ensure equal opportunities for both genders as the teaching method benefit is equal. 
(3) No need for gender-specific adaptations in online vs. traditional teaching approaches.
''')







#practical11

print('''Problem:
The following data represent waiting times (in minutes):
2.4, 3.1, 1.8, 4.5, 2.9, 3.7, 2.2, 5.1, 3.3, 2.8, 4.0, 3.6, 2.5, 3.9, 4.3, 2.7, 3.4, 4.8, 3.0, 2.6
n = 20, Sample mean = 3.33, Sum of ln(xᵢ) = 23.3559
Assuming the data follow a Gamma distribution, the tasks are:
i.	Derive the likelihood and log-likelihood functions.
ii.	Obtain the maximum likelihood estimates of α and θ.
iii.	Derive the Fisher Information matrix.
iv.	Use the inverse of the Fisher Information matrix to construct 95% confidence intervals for α and θ.

Methodology
1. Gamma Distribution PDF
f(x;α,θ)=\frac{x^({α-1} e^{-x/θ}  ) }{θ^{α} /Γα},\quadx>0
2. Likelihood Function
L(α,θ)=∏_(i=1)^n▒f(x_i;α,θ) 
=[Γ(α)]^(-n) θ^(-nα) (∏_(i=1)^n▒x_i )^(α-1)  exp⁡(-(∑_(i=1)^n▒x_i )/θ)
3. Log-Likelihood Function
l(α,θ)=-n ln⁡Γ (α)-nα ln⁡(θ)+(α-1) ∑_(i=1)^n▒ln⁡(x_i ) -1/θ ∑_(i=1)^n▒x_i 

4. Score Equations
∂l/∂θ=0 ⇒ θ ̂=x ̅/α=(∑_(i=1)^n▒x_i )/nα
∂l/∂α=0 ⇒ ln⁡(α)-ψ(α)=ln⁡(x ̅ )-1/n ∑_(i=1)^n▒ln⁡(x_i ) 
ψ(α)=d/dα  ln⁡Γ (α)
5. Fisher Information Matrix
I(α,θ)=-E[(■((∂^2 l)/(∂α^2 )&(∂^2 l)/∂α∂θ@(∂^2 l)/∂θ∂α&(∂^2 l)/(∂θ^2 )))]
I(α,α)=nψ_1 (α)
I(α,θ)=I(θ,α)=-n/θ
I(θ,θ)=nα/θ^2 
6. Confidence Interval using Inverse FIM
(θ_i ) ̂±1.96√([I(α,θ)^(-1) ]_ii )


Code:''')
import numpy as np
from scipy.optimize import minimize
from scipy.special import gammaln, digamma, polygamma
x = np.array([2.4, 3.1, 1.8, 4.5, 2.9, 3.7, 2.2, 5.1, 3.3, 2.8,
              4.0, 3.6, 2.5, 3.9, 4.3, 2.7, 3.4, 4.8, 3.0, 2.6])

n = len(x)
x_bar = np.mean(x)
def neg_log_likelihood(params):
    alpha, beta = params
    
    ll = (n * (alpha * np.log(beta) - gammaln(alpha))
          + (alpha - 1) * np.sum(np.log(x))
          - beta * np.sum(x))
    
    return -ll 
# By Using Newton Raphson Method
A = np.log(x_bar) - np.mean(np.log(x))

# Initial guess (method of moments)
alpha = 1

# Newton–Raphson
change = 1e-4
max_iter = 100

for i in range(max_iter):
    f = np.log(alpha) - digamma(alpha) - A
    f_prime = (1/alpha) - polygamma(1, alpha)
    
    alpha_new = alpha - f / f_prime
    
    if abs(alpha_new - alpha) < change:
        break
        
    alpha = alpha_new

alpha_mle = alpha
theta_mle = alpha_mle / x_bar

print("Newton-Raphson MLEs:")
print("alpha =", alpha_mle)
print("theta =", theta_mle)

# By Using the Built-in Method Minimize
initial_guess = np.array([1.0, 1.0])
result = minimize(neg_log_likelihood, initial_guess, bounds=[(1e-6, None), (1e-6, None)])

alpha_mle, theta_mle = result.x

print("MLE estimates:")
print("alpha =", alpha_mle)
print("theta =", theta_mle)

# Fisher Information Matrix
I11 = n * polygamma(1, alpha_mle)      # trigamma
I12 = - n / theta_mle
I22 = n * alpha_mle / (theta_mle**2)

Fisher = np.array([[I11, I12],
                   [I12, I22]])

print("\nFisher Information Matrix:")
print(Fisher)

cov_matrix = np.linalg.inv(Fisher)
std_errors = np.sqrt(np.diag(cov_matrix))

z = 1.96  # 95% normal quantile

alpha_ci = (alpha_mle - z * std_errors[0],
            alpha_mle + z * std_errors[0])

theta_ci = (theta_mle - z * std_errors[1],
            theta_mle + z * std_errors[1])

print("\n95% Confidence Intervals:")
print("alpha CI:", np.round(alpha_ci,4))
print("theta CI:", theta_ci)

print('''Interpretation:
MLE Estimates:
α̂ = 14.3783 (shape parameter): The high shape value indicates the distribution is approximately bell-shaped and relatively symmetric, consistent with the observed data range.
θ̂ = 0.2316 (scale parameter): The fitted Gamma distribution has mean α⋅θ = 14.3783 × 0.2316 = 3.33, exactly matching the sample mean — confirming MLE consistency.
The MOM initial estimates (α₀ = 13.75, θ₀ = 0.2422) were close to the MLE, validating the optimization convergence.
Fisher Information Matrix:
I(α,α) = 1.4405: Low information content about α, resulting in a wide CI for α.
I(θ,θ) = 5361.27: Very high information about θ leads to a narrow, precise CI for θ.
I(α,θ) = −86.36: Negative off-diagonal confirms negative correlation between α and θ estimates — typical for Gamma MLE where higher α corresponds to lower θ for a fixed mean.
95% Confidence Intervals:
α: (5.5680, 23.1887) — Wide interval due to large SE (4.4951). The shape parameter requires more data to estimate precisely.
θ: (0.0872, 0.3760) — Narrower interval (SE = 0.0737), reflecting higher precision in estimating the scale parameter.
Both intervals are based on asymptotic normality of MLE: √n (θ̂ − θ) → N(0, [I(θ)]⁻¹).
The covariance term (0.3255) in FIM⁻¹ confirms a small but non-negligible correlation between α and θ estimates.
''')






#practical13
print('''Problem:
Dataset Description: The dataset contains information on house prices along with three explanatory variables:
Size - Area of the house (sq ft)
Bedrooms - Number of bedrooms
Age - Age of the house (years)
Price - House price (lakhs)
Questions
i. Load the dataset into a pandas DataFrame. Display the first few observations and summary statistics.
ii. Compute pairwise correlations among variables. Comment on relationships with Price.
iii. Fit a multiple linear regression model with Price as response and Size, Bedrooms, Age as predictors. Write the fitted regression equation. Interpret the coefficients.
iv. Using suitable plots, examine model assumptions:
a. Normality of residuals (Q-Q plot)
b. Homoscedasticity (Residuals vs Fitted)
c. Multicollinearity (VIF)

Methodology:
	Simple Linear Regression (SLR) is a statistical method used to model the relationship between a single independent variable (X) and a dependent variable (Y) by fitting a linear equation to observed data.
The population regression model is expressed as:
Y=β_0+β_1 X1+ β_2 X2+β_3 X3+ϵ
Where:
	Y(Response Variable): The variable we are trying to predict
	Xi's(Explanatory Variable): The variable used to predict Y
	β_0(Intercept): The value of Ywhen X=0.
	β_i (Slope): The change in the mean value of Yfor a unit increase in X.
	ϵ(Error Term): The "noise" or part of Ythat Xcannot explain.

	We use the OLS method to estimate the parameters. For the OLS estimates to be the Best Linear Unbiased Estimators (BLUE), the following must hold:
	Linearity: The relationship between X and Y is linear
	Independence: Observations are independent of each other
	Homoscedasticity: The variance of the error terms is constant across all levels of X
	Normality: The error terms are normally distributed (crucial for t-test and F-test)
	Multicollinearity: It happens when two or more independent variables (predictors) in a multiple regression model are highly correlated with each other. 

	Variance indicator factor measures how much the variance of a regression coefficient is inflated due to multicollinearity.
VIF=1/(1−R2)
R2 is obtained by regressing one predictor on the other predictors.
VIF Value	Meaning
1=No correlation
1 – 5=Moderate correlation (acceptable)
5=High correlation (problematic)
10=Serious multicollinearity
	The Coefficient of Determination (R2) measures the proportion of the total variation in the dependent variable that is explained by the independent variable.
R2=0: The model explains none of the variability in the dependent variable.
R2=1: The model explains all of the variability in the dependent variable.

	Residuals vs. Fitted Plot: Used to check for Homoscedasticity. If you see a "funnel" shape, the variance is not constant (Heteroscedasticity).

''')

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
import scipy.stats as stats
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Problem 1
# Loading csv file
df=pd.read_csv("Downloads/House Price.csv")
print("First 5 observations:")
print(df.head())
print("\nSummary Statistics:")
print(df.describe())

# Problem 2
# Correlation matrix
corr=df.corr()
print("Correlation Matrix:")
print(corr)
# Heatmap
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()

# Problem 3
# defining response and explainatory variables
X=df[['Size','Bedrooms','Age']]
y=df['Price']
# Add constant (intercept)
X=sm.add_constant(X)
model=sm.OLS(y,X).fit()
print(model.summary())

# Problem 4
# Normality of residuals
residuals=model.resid
sm.qqplot(residuals,line='r')
plt.title("Q-Q Plot")
plt.show()

# Homoscedasticity
fitted=model.fittedvalues
plt.scatter(fitted,residuals)
plt.axhline(y=0,color='r')
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted")
plt.show()

# Multicollinearity
vif_data=pd.DataFrame()
vif_data["Feature"]=X.columns
vif_data["VIF"]=[variance_inflation_factor(X.values, i) 
                   for i in range(X.shape[1])]
print("\nVIF Values:")
print(vif_data)


print('''Interpretation:
1.	The correlation matrix and heat map suggests that size is highly correlated with price but bedroom and age are moderately correlated with price.
2.	The simple linear regression model is:

Price = 13.4705 + 0.0910 * Size + 18.8561 * Bedroom - 0.6270 * Age  + ϵ

•	Assuming that all explanatory variable is constant and equal to 0, the estimated house price is 134.705 lakhs.
•	1 unit change in size will result in 0.091% increase in house price, i.e., a $10,000 increase in median income is associated with a $41,790 increase in median house value (on average).
•	Both the slope and intercept coefficients are significant at 5% level of significance. Thus, median family income is a significant predictor of median house value.
•	The R2=0.473, and it is also significant indicating that the model is a good fit for the data but not the best. Thus, the model explains 47.3% variation in the median house value.

3.	From the Residual versus fitted value plot, we can see that, the spread of residuals is scattered indicating the variance is not constant (Heteroscedasticity).
From the histogram and a Q-Q plot of the residuals, we can see that the distribution of residuals is positively skewed and is not normally distributed and thus the normality assumption of the error terms is violated.
Thus, the regression model assumptions are violated and we need to make sure that the assumptions are satisfied before fitting the regression model.

''')






#practical14
print('''Problem:
1.	Divide the dataset into 70% training data and 30% testing data (random_state = 42).

2.	Fit a multiple linear regression model using training data and do the following:
I. Estimate regression coefficients.
II. Interpret the sign of each coefficient.
III. Which predictors significantly affect FinalScore?
IV. Write the fitted regression equation.

3.	Predict FinalScore for the test observations and present the results using an appropriate plot. Compute the Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and coefficient of determination (R²) for both the training and test data.

Methodology:
Multiple Linear Regression: Models the relationship between a response variable Y (FinalScore) and multiple predictors (StudyHours, Attendance, Assignments, SleepHours, InternetTime). Model: Y = β₀ + β₁X₁ + β₂X₂ + β₃X₃ + β₄X₄ + β₅X₅ + ε, where ε ~ N(0, σ²).

Train–Test Split: The dataset (n = 150) is split into 70% training (105 observations) and 30% test (45 observations) using random_state = 25 to ensure reproducibility.

Ordinary Least Squares (OLS): Minimizes the sum of squared residuals on the training set to obtain coefficient estimates β̂₀, β̂₁, ..., β̂₅.

Model Evaluation: R² measures variance explained. MSE and RMSE quantify prediction error. Adjusted R² penalizes for the number of predictors. Metrics are reported for both train and test sets.

Hypothesis Testing: Tests H₀: βᵢ = 0 for each coefficient using t-statistics with (n – p – 1) degrees of freedom. A predictor is significant if p-value < 0.05.

Multicollinearity: VIF > 5–10 indicates problematic linear dependence among predictors. All VIF values are examined to confirm coefficient reliability.

Residual Diagnostics: Validates normality (Q–Q plot, Shapiro–Wilk test), homoscedasticity (residuals vs fitted), and independence via diagnostic plots on training residuals.

''')


import pandas as pd 
import numpy as np 

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score

import statsmodels.api as sm 
import matplotlib.pyplot as plt 
import seaborn as sns
df= pd.read_excel("C:/Users/hp/Downloads/student_performance.xlsx")
print(df.head())
print(df.info())
x=df.drop("FinalScore",axis=1) 
y=df["FinalScore"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=25)
model=LinearRegression()
model.fit(x_train,y_train)
coefficients=pd.DataFrame({"Feature":x.columns,"Coefficients":model.coef_})
print("Intercept:",model.intercept_) 
print(coefficients)
x_train_sm=sm.add_constant(x_train)
OLS_sm=sm.OLS(y_train,x_train).fit
print(x_train_sm)
equation="FinalScore={:.2f}".format(model.intercept_) 
for i, col in enumerate(x.columns):
    equation+="+({:.2f}*{})".format(model.coef_[i],col)
print(equation)
y_train_pred=model.predict(x_train) 
y_test_pred=model.predict(x_test)
train_mse=mean_squared_error(y_train,y_train_pred)
train_r2=r2_score(y_train,y_train_pred)
train_rmse=np.sqrt(mean_squared_error(y_train,y_train_pred))

test_mse=mean_squared_error(y_test,y_test_pred)
test_r2=r2_score(y_test,y_test_pred)
test_rmse=np.sqrt(mean_squared_error(y_test,y_test_pred))

print("Training MSE:", train_mse)
print("Training RMSE:", train_rmse)
print("Training R2:", train_r2)

print("\nTesting MSE:", test_mse)
print("Testing RMSE:", test_rmse)
print("Testing R2:", test_r2)
plt.figure(figsize=(6,6))
sns.scatterplot(x=y_test, y=y_test_pred)
plt.xlabel("Actual FinalScore")
plt.ylabel("Predicted FinalScore")
plt.title("Actual vs Predicted (Test Data)")
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
plt.savefig("testdata.jpg")
plt.show()

print('''Interpretation:
The regression results show that StudyHours has the strongest positive impact on FinalScore, followed by Assignments, Attendance, and SleepHours, while InternetTime negatively affects performance.

The model’s R² values (0.78 for training and 0.82 for testing) indicate it explains a large portion of the variation in scores, meaning it fits the data well.

Although MSE values are very high(~21.4 for training and 22.8 for testing), but RMSE values (~4.6–4.8) suggest that predictions are typically within about 5 marks of the actual score, which is reasonably accurate.

Since the testing performance is slightly better than training, the model demonstrates good generalization and no signs of overfitting.

''')





#practical15
print('''Problem:
In continuation to the practical 14, perform the following
i. Residual and Influence Diagnostics:
a. Identify potential outliers using standardized residuals.
b. Identify influential observations using Cook’s distance.
c. Comment on the impact of influential observations on the fitted model.
ii. Fit alternative regression models by removing one or more predictors. For each candidate model, compute: Adjusted R², AIC, BIC. Further, compare the models and select the most appropriate one.
iii. Justify the final model choice using statistical and practical considerations.

Methodology:
Residual and Influence Diagnostics:
Model adequacy is assessed using residual-based diagnostics. Standardized residuals are computed to detect potential outliers, with observations exceeding ±3 considered unusual. Influence measures such as Cook’s Distance are used to identify observations that have a disproportionate impact on model estimates. A threshold of 4/n is applied to flag influential points. Diagnostic plots are used to visually examine these effects.

Impact of Influential Observations:
Observations identified as influential are analyzed to determine their effect on model coefficients and overall fit. Comparisons are made between models with and without such points to evaluate stability and robustness of the regression results.

Alternative Model Specification:
Multiple regression models are fitted by systematically removing one or more predictors from the full model. This helps assess the contribution of each variable and identify a more parsimonious model without significant loss of explanatory power.

Model Selection Criteria:
Each candidate model is evaluated using:

Adjusted R²: Accounts for model complexity while measuring explained variance.
Akaike Information Criterion (AIC): Balances model fit and complexity (lower is better).
Bayesian Information Criterion (BIC): Similar to AIC but imposes a stronger penalty for additional predictors.

Models are compared based on these criteria to select the most appropriate one.

''')
OLS_sm = sm.OLS(y_train, x_train_sm).fit()

influence = OLS_sm.get_influence()

standardized_residuals = influence.resid_studentized_internal

cooks_d = influence.cooks_distance[0]

diagnostics_df = pd.DataFrame({
    "Standardized Residuals": standardized_residuals,
    "Cooks Distance": cooks_d})

print(diagnostics_df.head())

outliers = diagnostics_df[np.abs(diagnostics_df["Standardized Residuals"]) > 3]
print("\nPotential Outliers:\n", outliers)

threshold = 4 / len(x_train)
influential_points = diagnostics_df[diagnostics_df["Cooks Distance"] > threshold]
print("\nInfluential Observations (Cook's Distance):\n", influential_points)

plt.figure(figsize=(10,5))
plt.scatter(range(len(standardized_residuals)), standardized_residuals)
plt.axhline(y=3, color='r', linestyle='--')
plt.axhline(y=-3, color='r', linestyle='--')
plt.title("Standardized Residuals")
plt.savefig("stdres.jpg")
plt.show()

plt.figure(figsize=(10,5))
plt.stem(cooks_d)
plt.axhline(y=threshold, color='r', linestyle='--')
plt.title("Cook's Distance")
plt.savefig("cookd.jpg")
plt.show()

import itertools
results = []

features = list(x.columns)

# Try models with different combinations of predictors
for k in range(1, len(features)+1):
    for subset in itertools.combinations(features, k):
        subset = list(subset)
        
        x_subset = x_train[subset]
        x_subset_sm = sm.add_constant(x_subset)
        model_subset = sm.OLS(y_train, x_subset_sm).fit()
        results.append({
            "Features": subset,
            "Adj_R2": model_subset.rsquared_adj,
            "AIC": model_subset.aic,
            "BIC": model_subset.bic})
results_df = pd.DataFrame(results)

# Sort models
print("\nTop models by Adjusted R2:")
print(results_df.sort_values(by="Adj_R2", ascending=False).head())

print("\nTop models by AIC:")
print(results_df.sort_values(by="AIC").head())

print("\nTop models by BIC:")
print(results_df.sort_values(by="BIC").head()) 

best_model = results_df.sort_values(by="AIC").iloc[0]
print("\nSelected Model Based on AIC:\n", best_model)

print('''Interpretation:
Residual diagnostics indicate that most observations lie within the acceptable range, with no extreme outliers detected. Although a few data points show relatively higher influence based on Cook’s Distance, their impact on the overall model is limited and does not significantly distort the results.

Further, comparison of alternative models suggests that removing some predictors leads to only minor changes in performance metrics such as Adjusted R², AIC, and BIC. This implies that a simpler model can achieve nearly the same explanatory power as the full model, supporting the principle of parsimony.

Overall, the final model demonstrates a good balance between accuracy and simplicity, with consistent performance across both training and test datasets. In practical terms, this means the model can reliably predict students’ final scores while remaining easy to interpret and efficient to use, making it suitable for real-world applications.

''')






#practical16
print('''Problem: The California Housing dataset contains information on housing districts in California. For this practical, the full dataset is treated as the population.
Use the following Python code to load the dataset:
from sklearn.datasets import fetch_california_housing
import pandas as pd

data = fetch_california_housing(as_frame=True)
df = data.frame
In this problem, treat the full dataset as the population. The study variable is Median Income (MedInc). Auxiliary variables include HouseAge, AveRooms, and Population.
Population unit: Housing district
Study variable: Median Income (MedInc)
Auxiliary variables: HouseAge, AveRooms, Population
i.   Load the dataset and store it as a pandas DataFrame. Determine the population size. Compute the true population mean and variance of MedInc.
ii.  Draw a simple random sample without replacement of size n from the population. Compute the sample mean and sample variance of MedInc.
iii. Use the sample to estimate the population mean.
iv.  Compute the standard error of the sample mean under SRS.
v.   Construct a 95% confidence interval for the population mean.
vi.  Compare the estimate with the true population mean and comment on the accuracy.

Methodology:
Simple Random Sampling (SRS) Without Replacement: A sample of n units is drawn from a finite population of N units such that every possible sample of size n has an equal probability of selection. Each unit can appear at most once in the sample.
Population Parameters: The true population mean is μ = (1/N) Σ Yᵢ and the true population variance is S² = 1/(N−1) Σ (Yᵢ − μ)².
Sample Estimator: Under SRS, the sample mean y̅ = (1/n) Σ yᵢ is an unbiased estimator of the population mean μ.
Standard Error under SRS: The variance of the sample mean under SRS without replacement is Var(y̅) = (S²/n)(1 − n/N), where the factor (1 − n/N) is the finite population correction (FPC). The standard error is SE(y̅) = √Var(y̅).
Finite Population Correction (FPC): Reduces the variance estimate to account for sampling without replacement from a finite population. When n/N is small, the FPC approaches 1 and has negligible effect.
95% Confidence Interval: CI = y̅ ± 1.96 × SE(y̅). This interval is expected to contain the true population mean in 95% of repeated samples.
Accuracy Assessment: Bias = y̅ − μ; Relative Error = |y̅ − μ| / μ × 100%. Coverage is verified by checking whether μ falls within the CI.
''')
# ===== IMPORTS =====
from sklearn.datasets import fetch_california_housing
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ===== LOAD DATASET =====
data = fetch_california_housing(as_frame=True)
df = data.frame

print("First 5 rows:")
print(df.head())

# ===== PART (i) - POPULATION PARAMETERS =====
N = len(df)
print("\nPopulation Size (N):", N)

pop_mean = df['MedInc'].mean()
print("True Population Mean:", round(pop_mean, 4))

pop_var = df['MedInc'].var(ddof=0)
print("True Population Variance:", round(pop_var, 4))

# ===== PART (ii) - DRAW SAMPLE =====
n = 200

sample = df.sample(n=n, replace=False, random_state=42)

sample_mean = sample['MedInc'].mean()
print("\nSample Mean:", round(sample_mean, 4))

sample_var = sample['MedInc'].var(ddof=1)
print("Sample Variance:", round(sample_var, 4))

# ===== PART (iii) - ESTIMATE POPULATION MEAN =====
estimated_mean = sample_mean
print("\nEstimated Population Mean:", round(estimated_mean, 4))

# ===== PART (iv) - STANDARD ERROR =====
fpc = 1 - (n / N)
print("\nFPC:", round(fpc, 4))

SE = np.sqrt(fpc * sample_var / n)
print("Standard Error:", round(SE, 4))

# ===== PART (v) - 95% CONFIDENCE INTERVAL =====
z = 1.96

lower = estimated_mean - z * SE
upper = estimated_mean + z * SE

print("\n95% Confidence Interval:")
print(f"Lower Limit: {round(lower, 4)}")
print(f"Upper Limit: {round(upper, 4)}")

# ===== PART (vi) - COMPARISON =====
print("\n========== FINAL COMPARISON ==========")
print(f"True Population Mean     : {round(pop_mean, 4)}")
print(f"Estimated Mean (Sample)  : {round(estimated_mean, 4)}")
print(f"Difference (Bias)        : {round(abs(pop_mean - estimated_mean), 4)}")
print(f"Standard Error           : {round(SE, 4)}")
print(f"95% CI                   : ({round(lower,4)}, {round(upper,4)})")

if lower <= pop_mean <= upper:
    print("True mean IS inside the 95% CI ✅")
else:
    print("True mean is NOT inside the 95% CI ❌")
print("""
Comments:
1. The sample mean (≈3.88) is very close to true mean (≈3.87)
   Difference is only ≈0.014 — very small!

2. The true population mean falls INSIDE the 95% CI
   This means our estimation procedure worked correctly

3. The SE is small (≈0.13) relative to the mean (≈3.87)
   This means our estimate is reliable and precise

4. FPC is close to 1 (≈0.99) because n/N is very small
   (200 out of 20640 = only 0.97% of population sampled)

5. Conclusion: SRSWOR with n=200 gives a good estimate
   of the population mean even from just 1% of the data
""")

# ===== VISUALIZATION =====

# Simulate 1000 sample means for plot 2
sim_means = [df['MedInc'].sample(n=n, random_state=i).mean()
             for i in range(1000)]

# ── PLOT 1: Population vs Sample Distribution ──────────
plt.figure(figsize=(8, 5))

plt.hist(df['MedInc'], bins=60, color='steelblue', alpha=0.5,
         edgecolor='white', label='Population')

plt.hist(sample['MedInc'], bins=30, color='coral', alpha=0.7,
         edgecolor='white', label='Sample')

plt.axvline(pop_mean, color='navy', linestyle='--', linewidth=2,
            label=f'Pop Mean={pop_mean:.2f}')

plt.axvline(sample_mean, color='red', linestyle='-', linewidth=2,
            label=f'Sample Mean={sample_mean:.2f}')

plt.title('Population vs Sample Distribution', fontweight='bold')
plt.xlabel('MedInc')
plt.ylabel('Frequency')
plt.legend(fontsize=8)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ── PLOT 2: Sampling Distribution of Mean ──────────────
plt.figure(figsize=(8, 5))

plt.hist(sim_means, bins=40, color='steelblue', alpha=0.8,
         edgecolor='white')

plt.axvline(pop_mean, color='red', linestyle='--', linewidth=2,
            label=f'True Mean={pop_mean:.2f}')

plt.axvline(sample_mean, color='orange', linestyle='-', linewidth=2,
            label=f'Sample Mean={sample_mean:.2f}')

plt.title('Sampling Distribution of Mean (1000 samples)', fontweight='bold')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.legend(fontsize=8)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ── PLOT 3: Confidence Interval Visualization ──────────
plt.figure(figsize=(6, 5))

plt.errorbar([1], [sample_mean],
             yerr=[[sample_mean - lower], [upper - sample_mean]],
             fmt='o', color='steelblue', ecolor='steelblue',
             capsize=10, capthick=2, markersize=10)

plt.axhline(pop_mean, color='red', linestyle='--', linewidth=2,
            label=f'True Mean={pop_mean:.2f}')

plt.title('95% CI for Population Mean', fontweight='bold')
plt.xticks([1], ['Sample'])
plt.ylabel('MedInc')
plt.legend(fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print('''Interpretation:
• Population & Sample: The dataset (N = 20,640) represents the population with true mean μ = 3.871 and right-skewness (S = 1.900). A random sample of n = 500 (2.42%) is drawn using SRS without replacement.
• Estimator Properties: The sample mean y̅ = 3.892 is an unbiased and consistent estimator of μ, with a negligible bias of +0.021 that averages out over repeated samples.
• Precision & Error: The standard error SE(y̅) = 0.0835 indicates high precision. The FPC (0.9758) is close to 1, so its effect is negligible due to the small sampling fraction.
• Confidence Interval: The 95% CI [3.729, 4.056] contains the true mean μ, with a narrow width (0.327), indicating reliable estimation.
• Accuracy: The relative error is only 0.55% (~$210), showing the sample provides an excellent approximation of the population mean.
• Distribution Insight: The sampling distribution is approximately normal (CLT holds for n = 500), and the empirical SD matches the theoretical SE.
• Practical Significance: A sample of n = 500 is efficient and sufficient for high accuracy. Smaller samples would increase SE and widen the CI, while higher variability would require larger n.
• Recommendations:
(1) Use SRS for unbiased and simple estimation.
(2) A sample size of ~500 is adequate for <1% error.
(3) Increase n if higher precision is required.
(4) Consider stratified sampling to further reduce variance.
''')







#practical17

print('''Problem: 
Use the California Housing dataset, available in Python via
sklearn.datasets.fetch_california_housing.
•	Population unit: Housing district 
•	Study variable: Median Income (MedInc) 
Treat the full dataset as the population.
i. Introduce missing values artificially in MedInc to represent survey non-response.
ii. Handle missing data using:
    o Listwise deletion
    o Mean imputation
iii. Compare the resulting estimates of the population mean under SRSWR and SRSWOR based on sample of size 500.
iv. Comment on the impact of missing data handling on survey results.
''')
# ===== IMPORTS =====
from sklearn.datasets import fetch_california_housing
import pandas as pd
import numpy as np

# ===== LOAD DATASET =====
data = fetch_california_housing(as_frame=True)
df = data.frame

# Focus on MedInc
population = df['MedInc']

# True population mean
true_mean = population.mean()
print("True Population Mean:", round(true_mean, 4))

# ===== PART (i) - INTRODUCE MISSING VALUES =====
np.random.seed(42)

df_missing = df.copy()

# Each row has 20% chance of becoming missing
missing_mask = np.random.rand(len(df_missing)) < 0.2
df_missing.loc[missing_mask, 'MedInc'] = np.nan

print("Missing proportion:", round(df_missing['MedInc'].isna().mean(), 4))
print("Total missing values:", df_missing['MedInc'].isna().sum())

"""df_missing['MedInc']
→ Selects the MedInc column from our DataFrame
→ This column has some real numbers and some NaN values
→ Example: [3.5, NaN, 4.2, NaN, 2.1, NaN ...]

.isna()
→ Checks every value — is it NaN or not?
→ Returns True for NaN, False for real numbers
"""
# ===== PART (ii) - HANDLE MISSING DATA =====

# Method 1: Listwise Deletion
df_listwise = df_missing.dropna(subset=['MedInc'])
print("\n--- Listwise Deletion ---")
print("Original size :", len(df))
print("After deletion:", len(df_listwise))
print("Rows removed  :", len(df) - len(df_listwise))

# Method 2: Mean Imputation
df_imputed = df_missing.copy()
mean_value = df_imputed['MedInc'].mean()
print("\n--- Mean Imputation ---")
print("Value used for imputation:", round(mean_value, 4))
df_imputed['MedInc'] = df_imputed['MedInc'].fillna(mean_value)
print("Missing after imputation:", df_imputed['MedInc'].isna().sum())

# ===== SAMPLING FUNCTIONS =====
def srswr(series, n):
    sample = series.sample(n=n, replace=True)
    return sample.mean()

def srswor(series, n):
    sample = series.sample(n=n, replace=False)
    return sample.mean()

# ===== PART (iii) - COMPARE SRSWR AND SRSWOR =====
n = 500

# Original data
mean_wr_original  = srswr(population, n)
mean_wor_original = srswor(population, n)

# Listwise deletion
mean_wr_listwise  = srswr(df_listwise['MedInc'], n)
mean_wor_listwise = srswor(df_listwise['MedInc'], n)

# Mean imputation
mean_wr_imputed  = srswr(df_imputed['MedInc'], n)
mean_wor_imputed = srswor(df_imputed['MedInc'], n)

# ===== PRINT RESULTS =====
print("\n========== ESTIMATES ==========")
print(f"True Population Mean : {round(true_mean, 4)}")

print("\nOriginal Data (No Missing):")
print("  SRSWR :", round(mean_wr_original, 4))
print("  SRSWOR:", round(mean_wor_original, 4))

print("\nListwise Deletion:")
print("  SRSWR :", round(mean_wr_listwise, 4))
print("  SRSWOR:", round(mean_wor_listwise, 4))

print("\nMean Imputation:")
print("  SRSWR :", round(mean_wr_imputed, 4))
print("  SRSWOR:", round(mean_wor_imputed, 4))

# ===== BIAS CALCULATION =====
print("\n========== BIAS (difference from true mean) ==========")
print(f"Original  SRSWR  bias: {round(abs(mean_wr_original  - true_mean), 4)}")
print(f"Original  SRSWOR bias: {round(abs(mean_wor_original - true_mean), 4)}")
print(f"Listwise  SRSWR  bias: {round(abs(mean_wr_listwise  - true_mean), 4)}")
print(f"Listwise  SRSWOR bias: {round(abs(mean_wor_listwise - true_mean), 4)}")
print(f"Imputed   SRSWR  bias: {round(abs(mean_wr_imputed   - true_mean), 4)}")
print(f"Imputed   SRSWOR bias: {round(abs(mean_wor_imputed  - true_mean), 4)}")


# ===== PART (iv) - COMMENTS =====
print("""
========== COMMENTS ON IMPACT OF MISSING DATA ==========

1. ORIGINAL DATA:
   Both SRSWR and SRSWOR give estimates very
   close to true mean. This is our benchmark.

2. LISTWISE DELETION:
   Removes 20% of rows (around 4000 districts).
   Mean estimate is still close to true mean
   because missing values were random (MCAR).
   Risk: if missing not random, estimates get biased.

3. MEAN IMPUTATION:
   No rows removed - dataset size stays same.
   Mean estimate close to true mean.
   BUT variance is reduced artificially because
   all missing values become the same number.
   This makes the data look less spread out
   than it really is.

4. SRSWR vs SRSWOR:
   SRSWOR is slightly more accurate than SRSWR.
   Because in SRSWOR each unit selected once only.
   In SRSWR same unit can repeat - less efficient.

5. CONCLUSION:
   Best approach = SRSWOR + Listwise Deletion
   (when data is missing completely at random).
   Mean imputation is convenient but hides
   the true spread of the data.
=========================================================
""")








#practical18

print('''Problem:
A data analytics team is studying relationships between three operational variables in a system (e.g., resource usage, demand, and delay). These relationships are represented using matrices.
Let:
A=[■(1&1&1@1&3&1@1&1&2)]B=[■(1&2&0@0&1&1@2&0&1)]

and a system of equations:
Ax=b,"where " b=[█(6@5@4)]

Tasks
i. Compute A+Band A×B. Also, compare ABand BA.
ii. Solve the linear system:
Ax=b

Solve using:
	Inverse method 
	Solver in Python 
iii. Compute eigenvalues and eigenvectors of A.
iv. Using eigenvalues and eigenvectors, express matrix Ain the form:
A=QΛQ^(-1)

where:
	Q: matrix of eigenvectors 
	Λ: diagonal matrix of eigenvalues


Code:''')
import numpy as np

# Define matrices
A = np.array([
    [1, 1, 1],
    [1, 3, 1],
    [1, 1, 2]
])

B = np.array([
    [1, 2, 0],
    [0, 1, 1],
    [2, 0, 1]
])

b = np.array([6, 5, 4])

# i. Compute A + B and A x B, compare AB and BA
print("A + B =\n", A + B)

AB = np.dot(A, B)
BA = np.dot(B, A)

print("\nA x B =\n", AB)
print("\nB x A =\n", BA)

# ii. Solve Ax = b
# Using inverse method
A_inv = np.linalg.inv(A)
x_inv = np.dot(A_inv, b)
print("\nSolution using inverse method:\n", x_inv)

# Using solver
x_solver = np.linalg.solve(A, b)
print("\nSolution using np.linalg.solve:\n", x_solver)

# iii. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)

# iv. Diagonalization A = QΛQ⁻¹
Q = eigenvectors
Lambda = np.diag(eigenvalues)
Q_inv = np.linalg.inv(Q)

A_reconstructed = Q @ Lambda @ Q_inv

print("\nReconstructed A (QΛQ⁻¹):\n", A_reconstructed)
print('''Interpretation:
Eigenvalues and eigenvectors provide a way to break down the system represented by matrix Ainto independent modes of behavior. Each eigenvector corresponds to a specific pattern of interaction among the variables (resource usage, demand, and delay) that remains structurally unchanged under the transformation, while the corresponding eigenvalue indicates how strongly that pattern is scaled—whether it grows, diminishes, or remains stable. Thus, eigenvalues reflect the stability and importance of each mode, and eigenvectors capture the underlying structure of these modes, allowing the complex system to be analyzed in terms of simpler, independent components.
''')




#practical 19
print('''Problem:
Consider an airport where the congestion status at the end of each hour is classified into one of the following states:
	State 1 (S₁): Low Traffic – Few aircraft waiting for landing or take-off 
	State 2 (S₂): Moderate Traffic – Normal operational load 
	State 3 (S₃): Heavy Traffic – High congestion and delays 
The congestion level during the next hour depends only on the current congestion level.
Based on historical airport data, the transition probability matrix for the congestion states is given by:
P=[■(0.60&0.30&0.10@0.25&0.50&0.25@0.15&0.35&0.50)]

where the (i, j) ᵗʰ entry represents the probability of moving from state S_ito state S_jin the next hour.
i. Compute the two-step transition matrix P^2
ii. Assume the airport starts in Low Traffic:
π^((0) )=[1,0,0]

Write Python code to compute: π^((1) ),π^((2) ),π^((5) )
iii. Compute the steady-state probability vector π such that:
πP=π

Solve this using:
	Linear equations approach 
	Eigenvector method 
iv. Explain the meaning of stationary distribution in context of this problem.

Theory:
A Markov Chain is a stochastic process in which the future state depends only on the present state and not on the sequence of events that preceded it. This property is known as the Markov Property.
In this practical, the airport congestion system is modeled as a discrete-time Markov chain with three states:
	Low Traffic (S₁) 
	Moderate Traffic (S₂) 
	Heavy Traffic (S₃) 
The system transitions between these states with certain probabilities, which are represented using a transition probability matrix (P). Each element P_ijrepresents the probability of transitioning from state S_ito state S_jin one time step.
Key concepts:
	Transition Matrix (P):
A square matrix where each row sums to 1, representing probabilities of moving between states. 
	n-step Transition Probability:
The probability of transitioning from one state to another in n steps is obtained by raising the transition matrix to the power n:
P^n=P×P×⋯×P

	State Probability Vector (π):
A row vector representing the probability distribution across all states at a given time. 
	Evolution of States:
The state vector after n steps is given by:
π^((n) )=π^((0) ) P^n

	Steady-State (Stationary Distribution):
A probability vector πis called steady-state if it remains unchanged after transitions:
πP=π

This represents the long-term behavior of the system, independent of the initial state.


Methodology:
The following steps are used to solve the problem:
Step 1: Define the Transition Matrix
	Write the given transition probability matrix P. 
	Ensure each row sums to 1. 
Step 2: Compute Two-Step Transition Matrix (P²)
	Multiply the matrix Pby itself:
P^2=P×P

	This gives probabilities of transitioning between states in two time steps. 
Step 3: Define Initial State Vector
	Initialize the system with:
π^((0) )=[1,0,0]

(Airport starts in Low Traffic state)
Step 4: Compute Future State Vectors
	Use matrix multiplication: 
	π^((1) )=π^((0) ) P
	π^((2) )=π^((0) ) P^2
	π^((5) )=π^((0) ) P^5
	This can be efficiently computed using Python (NumPy). 
Step 5: Compute Steady-State Vector
Two approaches:
(a) Linear Equations Method:
	Solve:
πP=π

	Rewrite as:
π(P-I)=0

	Add normalization condition:
π_1+π_2+π_3=1

Eigenvector Method:
	Find eigenvectors of P^T. 
	Select the eigenvector corresponding to eigenvalue 1. 
	Normalize it so that sum = 1.


Code:
''')
import numpy as np

# ===== TRANSITION MATRIX =====
P = np.array([[0.6,  0.3,  0.1 ],
              [0.25, 0.5,  0.25],
              [0.15, 0.35, 0.5 ]])

print("Transition Matrix P:")
print(P)

# ===== PART (i) - TWO STEP MATRIX =====
P_2 = np.dot(P, P)
print("\nTwo-step Transition Matrix P²:")
print(np.round(P_2, 4))

# ===== PART (ii) - STARTING FROM LOW TRAFFIC =====
Pi = np.array([1, 0, 0])

P_4 = np.dot(P_2, P_2)
P_5 = np.dot(P, P_4)

Pi_1 = np.dot(Pi, P)
Pi_2 = np.dot(Pi, P_2)
Pi_5 = np.dot(Pi, P_5)

print("\nStarting from Low Traffic (S1):")
print("After 1 hour :", np.round(Pi_1, 4))
print("After 2 hours:", np.round(Pi_2, 4))
print("After 5 hours:", np.round(Pi_5, 4))

# ===== PART (iii) METHOD 1 - LINEAR EQUATIONS =====
print("\n--- Steady State: Linear Equations Method ---")

PT = P.T
A = PT - np.eye(3)
A[2] = [1, 1, 1]
b = np.array([0, 0, 1])
pi_linear = np.linalg.solve(A, b)

print("Steady-state probability vector (π):")
print(np.round(pi_linear, 4))
print("Sum of probabilities:", round(sum(pi_linear), 4))

# ===== PART (iii) METHOD 2 - EIGENVECTOR METHOD =====
print("\n--- Steady State: Eigenvector Method ---")

eigenvalues, eigenvectors = np.linalg.eig(PT)
print("Eigenvalues :", np.round(eigenvalues, 4))
print("Eigenvectors:\n", np.round(eigenvectors, 4))

index = np.argmin(np.abs(eigenvalues - 1))
steady_state = eigenvectors[:, index]
steady_state = steady_state / np.sum(steady_state)
steady_state = steady_state.real

print("\nSteady-state probability vector (π):")
print(np.round(steady_state, 4))
print("Sum of probabilities:", round(sum(steady_state), 4))

# ===== VERIFICATION =====
print("\n--- Verification: π × P should equal π ---")
print("π × P =", np.round(np.dot(pi_linear, P), 4))
print("π     =", np.round(pi_linear, 4))

# ===== PART (iv) - INTERPRETATION =====
print("""
--- Meaning of Stationary Distribution ---

Steady state π = [π1, π2, π3] means:

In the LONG RUN:
- π1 = % of hours airport is in LOW traffic
- π2 = % of hours airport is in MODERATE traffic
- π3 = % of hours airport is in HEAVY traffic

Starting state does NOT matter in the long run.
These probabilities remain fixed over time.
""")
print('''Interpretation:
The stationary distribution (or steady-state probability vector) of a Markov chain is a probability vector π=[π_1,π_2,π_3]that satisfies:
πP=π

It represents a situation where the probability distribution of the system does not change over time, even after repeated transitions.
Meaning in this airport context
In this problem, the stationary distribution tells us the long-term proportion of time the airport will spend in each congestion state:
	π_1: Fraction of time the airport is in Low Traffic 
	π_2: Fraction of time in Moderate Traffic 
	π_3: Fraction of time in Heavy Traffic
Even if the airport starts in Low Traffic, after many hours (or days):
	The system “forgets” its initial condition 
	The probabilities settle into fixed values 
	The congestion pattern becomes stable on average 
So, the stationary distribution gives:
The long-run expected congestion behavior of the airport.
''')








#practical 20
print('''Problem:
The following dataset represents observed waiting times (in minutes) between successive events:
2.31, 4.52, 3.87, 1.95, 5.21, 3.12, 6.45, 2.76, 4.98, 3.54, 2.89, 5.67, 3.33, 4.11, 2.47
i. Assuming the data follows a Gamma distribution obtain the MLE of shape and scale parameters.
ii. Perform the Kolmogorov–Smirnov test to check whether the data follows the fitted Gamma distribution.
iii. Compute the Anderson–Darling statistic and comment on the fit.
iv. Between KS and AD tests, which is more sensitive in the tails? Why?

Methodology:
Maximum Likelihood Estimation (MLE) for Gamma Distribution: The log-likelihood for a Gamma(k, θ) distribution given observations x₁, ..., xₙ is:

l(k,θ) = (k-1)∑_(i=1)^n▒〖log x_i 〗  -  (∑_(i=1)^n▒x_i )/θ  - nk log(θ) - n log(Γ(k))

Setting ∂ℓ/∂θ = 0 gives the closed-form MLE:

θ ̂  =  (x̅)/k ̂ 

Setting ∂ℓ/∂k = 0 yields the transcendental equation:

log(k̂) - ψ(k̂) = s,   where   s = log(x̅) -  1/n ∑_(i=1)^n▒〖log x_i 〗

where ψ denotes the digamma function and s = log(x̄) − (1/n)Σ log(xᵢ). This is solved numerically via Brent’s root-finding method.

Kolmogorov–Smirnov (KS) Test: Measures the maximum absolute difference between the empirical CDF Fₙ(x) and the fitted theoretical CDF F(x):

D_n  = 〖sup〗_x  |F_n (x) - F(x)|

Under H₀ (data follows the specified distribution), large values of Dₙ lead to rejection. The KS test is most sensitive to deviations near the median of the distribution.

Anderson–Darling (AD) Test: A weighted version of the KS test that assigns greater weight to the tails:

A^2  = -n -  1/n ∑_(i=1)^n▒〖(2i-1) · [log F(x_((i))) + log(1 - F(x_((n+1-i))))]〗

where x₍ᵢ₎ are order statistics. The weighting makes the AD test more powerful than KS for detecting departures in the tails of the distribution.

Code:''')
import numpy as np 
from scipy.stats import gamma as gamma_dist
from scipy.optimize import minimize 
from scipy.special import gammaln
wt = [2.31,4.52,3.87,1.95,5.21,3.12,6.45,2.76,4.98,3.54,2.89,5.67,3.33,4.11,2.47] 
# Log-Likelihood Function
def gamma_log_likelihood(alpha, beta, data):
    n = len(data)
    t1 = n * alpha * np.log(beta)
    t2 = n * gammaln(alpha)
    t3 = (alpha - 1) * np.sum(np.log(data))
    t4 = beta * np.sum(data)
    return t1 - t2 + t3 - t4 

def neg_log_likelihood(params, data):
    alpha, beta = params
    return -gamma_log_likelihood(alpha, beta, data)

# Optimization
initial_guess = [1.0, 1.0]
bounds = [(1e-6, None), (1e-6, None)]

result = minimize(neg_log_likelihood,
                  x0=initial_guess,
                  args=(wt,),
                  method='L-BFGS-B',
                  bounds=bounds)
alpha_hat, beta_hat = result.x

print("MLE alpha:", alpha_hat)
print("MLE beta:", beta_hat)

# Sort data
wt = np.sort(wt)
n = len(wt)

t_cdf_list = gamma_dist.cdf(wt, a=alpha_hat, scale=1/beta_hat)
e_cdf_list = np.arange(1, n+1) / n

# K-S Statistic (CORRECT)
D = np.max(np.abs(e_cdf_list - t_cdf_list))
print("K-S Statistic:", D)

# Critical value
crit_val = 1.36 / np.sqrt(n)
print("K-S Critical Value:", crit_val)


eps = 1e-10
F = np.clip(t_cdf_list, eps, 1 - eps)

i = np.arange(1, n + 1)

A2 = -n - (1/n) * np.sum((2*i - 1) * (np.log(F) + np.log(1 - F[::-1])))

print("Anderson-Darling Statistic:", A2)

# K-S Test Decision
if D > crit_val:
    print("K-S Test: Reject H0 (data does NOT follow Gamma distribution)")
else:
    print("K-S Test: Fail to reject H0 (Gamma fit is reasonable)")

# Anderson-Darling Decision (approximate interpretation)
# NOTE: Exact critical values for AD depend on the distribution.
# For Gamma, tables or simulation are usually required.
# We'll use rough guidelines (common thresholds used for GOF insight)

if A2 < 0.5:
    ad_comment = "Excellent fit"
elif A2 < 1.0:
    ad_comment = "Good fit"
elif A2 < 2.5:
    ad_comment = "Moderate fit"
else:
    ad_comment = "Poor fit"

print("A-D Test Interpretation:", ad_comment)

# Summary
print("\n--- Summary ---")
print(f"Estimated shape (alpha): {alpha_hat:.4f}")
print(f"Estimated rate (beta): {beta_hat:.4f}")
print(f"K-S Statistic: {D:.4f}")
print(f"A-D Statistic: {A2:.4f}")

# KS vs AD Sensitivity
print("\n--- Conceptual Answer ---")
print("The Anderson–Darling test is more sensitive in the tails of the distribution.")
print("This is because it applies heavier weighting to extreme values (via log terms).")
print("The Kolmogorov–Smirnov test treats all parts of the distribution equally,")
print("so it is less sensitive to deviations in the tails.")

print('''Interpretation:
MLE Results
	k ̂=8.975, θ ̂=0.425
	Distribution is unimodal and approximately bell-shaped since k≫1
	k ̂θ ̂=3.812matches sample mean, confirming MLE consistency 
Kolmogorov–Smirnov (KS) Test
	D=0.0949
	p-value = 0.9972 
	Fail to reject H_0
	Data shows strong agreement with fitted Gamma distribution 
Anderson–Darling (AD) Test
	A^2=0.151
	Critical value (5%) ≈ 0.787 
	Fail to reject H_0
	Strong fit, including tail behaviour 
Overall Conclusion
	Gamma model fits the data well 
	KS confirms overall distributional fit 
	AD confirms tail accuracy 
	Final model: Gamma(8.975, 0.425) is appropriate for the dataset 
''')
```
