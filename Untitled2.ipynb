```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Load dataset
data = fetch_california_housing(as_frame=True)
df = data.frame

# i. First few observations and summary statistics
print("First 5 observations:")
print(df[['MedHouseVal', 'MedInc']].head())
print("\nSummary Statistics:")
print(df[['MedHouseVal', 'MedInc']].describe())

# ii. Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df['MedInc'], df['MedHouseVal'], alpha=0.3, color='steelblue', s=10)
plt.xlabel("Median Income (MedInc)")
plt.ylabel("Median House Value (MedHouseVal)")
plt.title("Scatter Plot: MedHouseVal vs MedInc")
plt.tight_layout()
plt.savefig("scatter.jpg")
plt.show()

# iii. Fit Simple Linear Regression
X = df[['MedInc']].values
y = df['MedHouseVal'].values

model = LinearRegression()
model.fit(X, y)

b0 = model.intercept_
b1 = model.coef_[0]

print(f"\nEstimated Regression Equation:")
print(f"MedHouseVal = {round(b0, 4)} + {round(b1, 4)} * MedInc")
print(f"\nInterpretation: For every 1 unit increase in Median Income,")
print(f"Median House Value increases by {round(b1, 4)} units on average.")

# iv. Hypothesis test for slope (H0: B1=0, H1: B1≠0)
n = len(y)
y_pred = model.predict(X)
residuals = y - y_pred

SSE = np.sum(residuals ** 2)
SSX = np.sum((df['MedInc'] - df['MedInc'].mean()) ** 2)

s2 = SSE / (n - 2)                    # MSE
se_b1 = np.sqrt(s2 / SSX)            # Standard error of slope
t_stat = b1 / se_b1                   # t statistic
p_val = 2 * (1 - stats.t.cdf(abs(t_stat), df=n - 2))  # two-tailed p-value

print(f"\nHypothesis Test for Slope (H0: β1 = 0):")
print(f"t-statistic : {round(t_stat, 4)}")
print(f"p-value     : {round(p_val, 6)}")
if p_val < 0.05:
    print("Conclusion: MedInc IS a statistically significant predictor of MedHouseVal (reject H0)")
else:
    print("Conclusion: MedInc is NOT a statistically significant predictor (fail to reject H0)")

# v. R-squared
r2 = r2_score(y, y_pred)
print(f"\nR² Value: {round(r2, 4)}")
print(f"Interpretation: {round(r2*100, 2)}% of the variation in MedHouseVal")
print(f"is explained by MedInc.")

# vi. Residuals vs Fitted Values plot
plt.figure(figsize=(8, 5))
plt.scatter(y_pred, residuals, alpha=0.3, color='darkorange', s=10)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.tight_layout()
plt.savefig("residuals_vs_fitted.jpg")
plt.show()
print("\nIf residuals are randomly scattered around 0 with no clear pattern,")
print("linearity and constant variance (homoscedasticity) assumptions are satisfied.")

# vii. Histogram and Q-Q plot of residuals
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histogram
axes[0].hist(residuals, bins=50, edgecolor='black', color='steelblue')
axes[0].set_xlabel("Residuals")
axes[0].set_ylabel("Frequency")
axes[0].set_title("Histogram of Residuals")

# Q-Q Plot
stats.probplot(residuals, dist="norm", plot=axes[1])
axes[1].set_title("Q-Q Plot of Residuals")

plt.tight_layout()
plt.savefig("residual_plots.jpg")
plt.show()
print("\nIf histogram is bell-shaped and Q-Q plot points lie on the diagonal line,")
print("the normality assumption of error terms is satisfied.")
```
