```python
import numpy as np 
import pandas as P
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

#Set random seed for for reproducibility 
np.random.seed(51)
size=500

#Normal,Exponential,Uniform,Poisson 
data={
      "Normal":
      np.random.normal(loc=0,scale=1,size=size),
      "Exponential":
      np.random.exponential(scale=1,size=size),
      "Uniform":
      np.random.uniform(low=0,high=1,size=size),
      "Poisson":
      np.random.poisson(lam=5,size=size)
     }

#Using pandas 
import pandas as pd
df=pd.DataFrame(data)
p_summary=df.describe()
print(p_summary)

#Using numpy 
numpy_summary = pd.DataFrame({
    "Mean": np.mean(df, axis=0),
    "Median": np.median(df, axis=0),
    "Variance": np.var(df, axis=0, ddof=1),
    "Std_dev": np.std(df, axis=0, ddof=1),
    "Min": np.min(df, axis=0),
    "Max": np.max(df, axis=0),
    "Skewness": df.skew(),
    "Kurtosis": df.kurt()
})
print(numpy_summary)
for name, value in data.items():
    plt.figure() 

    # Using Numpy
    plt.hist(value,bins=30,density=True,alpha=0.3,label="Numpy")

    #Using Seaborn
    sns.histplot(value,bins=25,stat="density",label="Seaborn")
    #Create X values for smooth curve
    x=np.linspace(min(value),max(value),500) 
    #Plot theoretical PDF/PMF
    if name=="Normal":
        plt.plot(x,stats.norm.pdf(x,0,1),label="PDF")
    elif name=="Exponential": 
        plt.plot(x,stats.expon.pdf(x),label="PDF")
    elif name=="Uniform": 
        plt.plot(x,stats.uniform.pdf(x,0,1),label="PDF")
    elif name=="Poisson": 
        plt.plot(x,stats.poisson.pmf(np.round(x),4),label="PMF")
    plt.legend()
    plt.title(name+" Distribution")
    plt.savefig("histda.jpg")
    plt.show()
plt.figure(figsize=(12, 10))
i = 1
for name, data in data.items():
    plt.subplot(2, 2, i)
    stats.probplot(data, dist="norm", plot=plt)
    plt.title(f"Q-Q Plot : {name}")
    i += 1
plt.tight_layout()
plt.show()

df["log_transformed"]=np.log(df["Exponential"])
df["Standardised"]=(df["log_transformed"]-df["log_transformed"].mean())/df["log_transformed"].std() 
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
stats.probplot(df["Exponential"],dist="norm",plot=plt)
plt.title("Q-Q Plot Original Exponential Data")
plt.subplot(1,2,2) 
stats.probplot(df["Standardised"],dist="norm",plot=plt) 
plt.title("Q-Q Plot Log-Transformed & Standardised Exponential Data")
plt.tight_layout()
plt.savefig("Std_sub.jpg")
plt.show()
```
