```python
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
product_A=[120,135,150,160,170,180,175,165,155,145,140,130]
product_B=[100,110,115,130,140,150,160,155,145,135,125,115]
product_C=[90,95,100,105,110,120,125,130,128,120,110,100]
product_D=[80,85,90,95,100,105,110,108,102,95,90,85]
```

```python
a=plt.bar(months,product_A)
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('bar_plot2.jpg')
plt.show()
```

```python
plt.figure()
plt.bar(months, product_A, color='steelblue')
plt.xlabel('Month')
plt.ylabel('Sales Units')
plt.title('Monthly Sales of Product A')
plt.show()
```

```python
plt.hist(product_B)
plt.savefig('hist_plot1.jpg')
plt.show()

plt.hist(product_B,bins=4,edgecolor='black',color='coral')
plt.xlabel('Sales units')
plt.ylabel('Frequency')
plt.title('Histogram of Product B Sales')
plt.savefig('hist_plot2.jpg')
```

```python
print('Total annual sales for product A:',sum(product_A))
print('Total annual sales for product B:',sum(product_B))
print('Total annual sales for product C:',sum(product_C))
print('Total annual sales for product D:',sum(product_D))

#b
#to plot a pie chart showing the percentage contribution of each product to the total sales 
#display percentage values on the chart

total_A = sum(product_A)
total_B = sum(product_B)
total_C = sum(product_C)
total_D = sum(product_D)

totals = [total_A, total_B, total_C, total_D]
labels = ['Product A','Product B','Product C','Product D']


plt.pie(totals, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Annual Sales Contribution by Product')
plt.show()
```

```python
plt.plot(months, product_B, marker='o', color='green')
plt.xlabel('month')
plt.ylabel('sales')
plt.title('Monthly sales trend of Product B')
plt.show()
```

```python
#to make a multiple line chart 
plt.plot(product_A,marker='o',label='Product A')
plt.plot(product_B,marker='d',label='Product B')
plt.plot(product_C,marker='s',label='Product C')
plt.plot(product_D,marker='^',label='Product D')
plt.xlabel('Months')
plt.ylabel('Sales Units')
plt.title('Comapring of monthly sales trends')
plt.legend()
plt.savefig('multiline3.jpg')
plt.show()
```
